```python
numeric_colums = df.select_dtypes(include=['number'])
corr_sale_price=numeric_colums.corr()["SalePrice"]
print(corr_sale_price.sort_values(ascending=False).head(8))

# Vizual koramiz:
import seaborn as sns
cols = corr_sale_price.sort_values(ascending=False).head(8).index
sns.heatmap(numeric_colums[cols].corr(),annot=True)
```

1. Kod nima qilyapti?

cols = corr_sale_price.sort_values(ascending=False).head(8).index — Uy narxiga (SalePrice) eng kuchli ta'sir qiluvchi (bog'langan) eng yaxshi 8 ta ustunni tanlab olyapti.
sns.heatmap(..., annot=True) — O'sha tanlangan 8 ta ustunning o'zaro va uy narxi bilan bog'liqligini rangli jadval (heatmap) shaklida chizib beryapti. annot=True raqamlarning o'zini ham kataklar ichiga yozib qo'yishni ta'minlaydi.

2. Natijalar nimani bildiradi? (Raqamlar tahlili)

Jadvalning eng birinchi ustuni / qatori bu — SalePrice (Uyning narxi). Narxga qaysi omillar eng ko'p ta'sir qilishini shu yerdan ko'ramiz:

**OverallQual (0.79): Uyning umumiy sifati va bezagi.**
Narxga eng kuchli ta'sir qiluvchi omil shu (79% bog'liqlik). Uyning sifati qancha baland bo'lsa, narxi shuncha qimmat bo'ladi.

**GrLivArea (0.71): Uyning yashash maydoni (kvadrat metri)**. Maydon qanchalik katta bo'lsa, narx shuncha yuqori bo'ladi.

**GarageCars (0.64) va GarageArea (0.62): Garajga sig'adigan mashinalar soni va garajning umumiy hajmi.** Garaj qancha katta bo'lsa, uy narxi shuncha qimmatlashadi.

**TotalBsmtSF (0.61) va 1stFlrSF (0.60): Zirza (podval) maydoni va birinchi qavatning maydoni.**

3. Heatmap ranglarining ma'nosi:

To'q qizg'ish / och pushti ranglar (1.0 ga yaqin): Ikki ustun bir-biriga juda kuchli bog'langanini bildiradi. Masalan, GarageCars bilan GarageArea o'rtasidagi bog'liqlik 0.88 — ya'ni garaj maydoni katta bo'lsa, unga sig'adigan mashina soni ham ko'p bo'ladi (bu mantiqiy).
To'q jigarrang / qora ranglar (0.3 - 0.4 atrofida): Bog'liqlik unchalik yuqori emasligini bildiradi.

🎯 Xulosa:
Chiziqli regressiya (Linear Regression) modelini o'qitayotganda, biz mana shu eng yuqori korrelyatsiyaga ega ustunlarni (Features) tanlab olsak, modelimiz yanadachaqqon va aniqroq ishlaydi!

```python
# topshiriq 1
# GrLivArea  va SalePrice orasida scatterplot chizish va galati nuhtalarni topish

sns.scatterplot(x=df["GrLivArea"], y=df["SalePrice"])

# Grafik chiroyli chiqishi uchun nomlar beramiz (ixtiyoriy)
plt.title("Yashash maydoni va Uy narxi bog'liqligi")
plt.xlabel("GrLivArea (Maydon)")
plt.ylabel("SalePrice (Narx)")

plt.show()
```

Hozir biz faqat GrLivArea (umumiy yashash maydoni) va SalePrice (narx) ustunlarini olib, 2 o'lchamli (2D) oddiy grafikda ko'ryapmiz. Shuning uchun bu yerda uyning boshqa xususiyatlari ko'rinmayapti.


Savolingizga javob: Ha, albatta ahamiyati bor! Hozir bu ikkita g'alati nuqta pastda qolib ketishining sababi aynan boshqa omillarning hisobga olinmaganidadir. Buni quyidagicha tushuntirish mumkin:

Uyning sifati (OverallQual): Faraz qilaylik, uylardan birining maydoni juda katta (masalan, 4500 kv.m), lekin u juda eski, ta'mirtalab, eshik-derazalari singan yoki yarim chala qurilgan uy bo'lishi mumkin. Maydoni katta bo'lgani bilan sifati past bo'lgani uchun uni arzon sotishgan. Chiziqli regressiya esa buni bilmaydi, u faqat "maydoni kattami — demak qimmat bo'lishi kerak" deb o'ylaydi va shuning uchun bu uy unga "anomaliya" (g'alati nuqta) bo'lib ko'rinadi.

Joylashuvi (Location / Neighborhood): Katta maydonli uy shaharning chekkasida, xarobaroq yoki qulay bo'lmagan joyda qurilgan bo'lishi mumkin. Maydon katta, lekin joylashuvi yomon bo'lgani uchun narxi past.

Qurilish turi yoki holati: Ba'zi uylar kim oshdi savdosida (bankrot bo'lgan egasining uyi) juda arzon narxda sotib yuborilgan bo'lishi mumkin.

💡 Xulosa:
Hozircha biz oddiy (bitta o'zgaruvchili) chiziqli regressiya qilyapmiz, shuning uchun faqat maydonga qarab narxni taxmin qilmoqchimiz. Lekin haqiqiy hayotda va keyingi qadamda ko'p o'zgaruvchili regressiyaga (Multiple Linear Regression) o'tganda, modelimizga faqat maydonni emas, balki uyning sifati, yili, garaji va boshqa ustunlarni ham qo'shamiz. Shunda bu "g'alati nuqtalar" o'z-o'zidan mantiqiy joyiga tushadi (chunki model "ha, maydoni katta ekan, lekin sifati 10 balldan 3 ekan, shuning uchun arzon ekan" deb tushunib yetadi).

```python
# end 60 foizdan oshiq bosh qiymatli ustunlarni topib ulanu royhatdan chiqarish va mechta chiqanin topsh
# 1. Har bir ustundagi bo'sh qiymatlar foizini hisoblaymiz
null_percentage = df.isnull().sum()/len(df)*100

# 2. 60% dan ko'p bo'sh qiymati bor ustunlarni tanlab olamiz
cols_to_drop = null_percentage[null_percentage > 60].index

# 3. Nechta shunday ustun borligini chiqaramiz
print(f" 60% dan ortiq bo'sh qiymatli ustunlar soni: {len(cols_to_drop)} ta")
print("O'chiriladigan ustunlar ro'yxati:")
print(cols_to_drop.tolist())

# 4. Ularni asosiy datasetdan o'chirib tashlaymiz
df = df.drop(columns=cols_to_drop)

print(f"\n Datasetning yangi o'lchami (qator, ustun): {df.shape}")
```
