```python
import numpy as np
import matplotlib.pyplot as plt

# Funksiya va uning hosilasi
def f(x):
    return x**2

def df(x):
    return 2 * x

x = np.linspace(-3, 3, 400)

plt.figure(figsize=(10, 5))

# 1. Asosiy funksiya f(x) = x^2
plt.plot(x, f(x), color='blue', linewidth=2.5, label="Funksiya: f(x) = x²")

# 2. x = 1 Nuqtadagi urinma (Tangent Line) chizamiz
# Urinma tenglamasi: y = f(x0) + f'(x0) * (x - x0)
x0 = 1.0
y0 = f(x0)
slope = df(x0) # slope = 2.0

x_tangent = np.linspace(0, 2, 100)
y_tangent = y0 + slope * (x_tangent - x0)

plt.plot(x_tangent, y_tangent, color='red', linestyle='--', linewidth=2, label=f"x={x0} dagi urinma (Qiyalik/Hosila = {slope})")
plt.scatter([x0], [y0], color='red', s=70, zorder=5)

# 3. x = 0 (Minimum nuqta) dagi hosila=0 ni ko'rsatish
plt.scatter([0], [0], color='green', s=70, zorder=5)
plt.axhline(0, color='green', linestyle=':', label="x=0 dagi hosila = 0 (Minimum)")

plt.title("f(x) = x² va uning nuqtadagi Hosilasi (Urinma qiyaligi)")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. CHAIN RULE HISOBLASH
# y = (x^2 + 1)^3
# u = x^2 + 1  => du/dx = 2x
# y = u^3      => dy/du = 3u^2
# dy/dx = (dy/du) * (du/dx) = 3(x^2 + 1)^2 * 2x

def forward_pass(x):
    u = x**2 + 1
    y = u**3
    return u, y

def chain_rule_backward(x, u):
    dy_du = 3 * (u**2)
    du_dx = 2 * x
    dy_dx = dy_du * du_dx  # Zanjir ko'paytmasi
    return dy_dx

x_val = 2.0
u_val, y_val = forward_pass(x_val)
grad = chain_rule_backward(x_val, u_val)

print(f"--- CHAIN RULE RESULT ---")
print(f"x = {x_val} nuqtada y = {y_val}")
print(f"Zanjir qoidasi bo'yicha hosila dy/dx = {grad}")

# 2. OPTIMIZATSIYA (GRADIENT DESCENT)
def loss_fn(w):
    return w**2

def loss_grad(w):
    return 2 * w

w = 4.0        # Boshlang'ich nuqta
lr = 0.2       # Learning rate
w_history = [w]
loss_history = [loss_fn(w)]

for step in range(10):
    g = loss_grad(w)
    w = w - lr * g  # Optimizatsiya qadami
    w_history.append(w)
    loss_history.append(loss_fn(w))

# GRAFIKLARNI CHIZISH
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# 1-Grafik: Murakkab Funksiya
x_range = np.linspace(-2.5, 2.5, 200)
u_range = x_range**2 + 1
y_range = u_range**3
ax1.plot(x_range, y_range, label='$y = (x^2 + 1)^3$', color='blue', linewidth=2)
ax1.scatter([x_val], [y_val], color='red', s=70, zorder=5, label=f'Nuqta x={x_val}')
ax1.set_title("Murakkab Funksiya Grafigi", fontsize=12)
ax1.set_xlabel("x (Kirish)")
ax1.set_ylabel("y (Chiqish)")
ax1.grid(True)
ax1.legend()

# 2-Grafik: Gradient Descent
w_range = np.linspace(-4.5, 4.5, 100)
ax2.plot(w_range, loss_fn(w_range), color='purple', alpha=0.7, label='Loss funksiyasi $L(w) = w^2$')
ax2.plot(w_history, loss_history, 'o--', color='red', linewidth=2, markersize=6, label='Optimizatsiya qadamlari')
ax2.set_title("Gradient Descent Orqali Optimizatsiya", fontsize=12)
ax2.set_xlabel("Vazn (w)")
ax2.set_ylabel("Xatolik (Loss)")
ax2.grid(True)
ax2.legend()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Boshlang'ich ma'lumotlar va parametrlarni beramiz
x = 2.0         # Kirish qiymati (Input)
y_true = 10.0   # Maqsaddagi haqiqiy natija (Target)

# Vaznlar (Weights) - dastlab ixtiyoriy sonlar
w1 = 1.5
w2 = 2.0
learning_rate = 0.05

history_loss = []
history_w1 = []

# 2. Optimization (Modelni o'qitish sikli)
for epoch in range(20):
    # FORWARD PASS (Oldinga o'tish: f(g(x)))
    g = w1 * x          # 1-qatlam chiqishi
    y_pred = w2 * g     # 2-qatlam (oxirgi bashorat)

    # Xatolikni hisoblash (MSE - Mean Squared Error)
    loss = (y_pred - y_true) ** 2

    history_loss.append(loss)
    history_w1.append(w1)

    # BACKPROPAGATION (Chain Rule yordamida hosilalar olish)
    # dLoss/dy_pred = 2 * (y_pred - y_true)
    dloss_dypred = 2 * (y_pred - y_true)

    # dypred/dw2 = g
    dypred_dw2 = g

    # dypred/dg = w2
    dypred_dg = w2

    # dg/dw1 = x
    dg_dw1 = x

    # CHAIN RULE: dLoss/dw1 = (dLoss/dypred) * (dypred/dg) * (dg/dw1)
    dloss_dw1 = dloss_dypred * dypred_dg * dg_dw1
    dloss_dw2 = dloss_dypred * dypred_dw2

    # OPTIMIZATION (Vaznlarni yangilash)
    w1 = w1 - learning_rate * dloss_dw1
    w2 = w2 - learning_rate * dloss_dw2

# 3. Natijalarni grafikda tasvirlash
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(history_loss, color='red', marker='o')
plt.title("Xatolikning Kamayishi (Loss)")
plt.xlabel("Qadamlar (Epochs)")
plt.ylabel("Xatolik Qiymati")
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(history_w1, color='green', marker='o')
plt.title("w1 Vaznining Optimallashuvi")
plt.xlabel("Qadamlar (Epochs)")
plt.ylabel("w1 qiymati")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. FUNKSIYANI TANIQTIRAMIZ (Retsept)
def calculate_y(x):
    u = 2 * x - 3
    y = u ** 2
    return u, y

# 2. MA'LUMOTLAR RO'YXATI (List)
x_list = [4, 0, 1,5]

# Natijalarni saqlash uchun ro'yxatlar
u_results = []
y_results = []

print("--- FUNKSIYA ORQALI HISOBLASH ---")
for x in x_list:
    # Funksiyani chaqiramiz
    u, y = calculate_y(x)

    u_results.append(u)
    y_results.append(y)

    print(f"Kirish x = {x}  -->  Funksiya natijasi: u = {u}, y = {y}")

# GRAFIKNI CHIZISH
plt.figure(figsize=(8, 5))

# Asosiy uzluksiz funksiya grafigi
x_continuous = np.linspace(-1, 5, 100)
_, y_continuous = calculate_y(x_continuous)
plt.plot(x_continuous, y_continuous, color='dodgerblue', linestyle='--', label='$y = (2x - 3)^2$ funksiya grafigi')

# Funksiyadan chiqqan nuqtalar
plt.scatter(x_list, y_results, color='red', s=100, zorder=5, label='Funksiyadan chiqqan nuqtalar')

for i in range(len(x_list)):
    plt.annotate(f"x={x_list[i]} -> y={y_results[i]}",
                 (x_list[i], y_results[i]),
                 textcoords="offset points",
                 xytext=(0, 10),
                 ha='center',
                 fontweight='bold')

plt.title("Funksiya (def) va List Bilan Hisoblangan Grafik", fontsize=12)
plt.xlabel("x (Kirish)")
plt.ylabel("y (Chiqish)")
plt.grid(True)
plt.legend()
plt.show()
```
