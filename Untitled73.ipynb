```python
df = df.drop(columns=["Cabin"])
df = df.dropna(subset=["Embarked"])
df["Age"] = df["Age"].fillna( df["Age"].median())
```

```python
print(df.shape)
df.isnull().sum()
```

```python
df["family_size"] = df['SibSp'] +df["Parch"] +1
df["is_alone"] = df["family_size"] == 1
print(df["family_size"].describe())
print(df["is_alone"].value_counts())
```

```python
#title dgan yangi ustuni yaratik Name degan ustini ichdegi malumotlarni shu title ichga tiqamiz
df["Title"] = df["Name"].str.extract(r",\s*([^\.]+)\.")
print(df["Title"].value_counts())
```

1. Regex (Regular Expression) qismi: r",\s*([^\.]+)\."
Bu kod Name ustunidan odamlarning unvonini (titulini) ajratib oladi.
r",\s*": Verguldan keyin kelgan bo'sh joylarni topadi (masalan: "Braund, Mr..." qismidagi vergul va undan keyingi joy).
([^\.]+): Bu qism unvonni ushlab oladi.
[^ \.] degani "nuqta bo'lmagan har qanday belgi" degani.
+ degani "bir yoki undan ortiq belgi" degani.
Ya'ni, bu nuqtagacha bo'lgan barcha harflarni ("Mr", "Mrs", "Miss", "Master" kabi) oladi.
\.: Unvondan keyin keladigan nuqtani topadi (titul tugaganini bildiradi).

```python
# 1. Titleda kam uchriydigan unvonlarni tanlab olamiz
kam_unvonlar = [
    "Dr", "Rev", "Col", "Mlle", "Major", "Ms", "Mme",
    "Don", "Lady", "Sir", "Capt", "the Countess", "Jonkheer"
]

# 2. 'Title' ustunidagi unvonlarni 'Rare' ga o'zgartiramiz
df["Title"] = df["Title"].replace(kam_unvonlar, "Rare")

# 3. Natijani tekshiramiz
print(df["Title"].value_counts())
```

```python
df["fare_per_person"] = df["Fare"] / (df["family_size"])
print(df["fare_per_person"].describe())
print(df[["Name","Fare","family_size","fare_per_person"]].head(2))
```

```python
# 'fare_per_person' ni guruhlarga bo'lish (masalan, arzon va qimmat chiptalar)
df['fare_category'] = pd.qcut(df['fare_per_person'], q=4, labels=['Low', 'Medium', 'High', 'Very High'])

# Qaysi kategoriya ko'proq omon qolgan?
print(df.groupby('fare_category')['Survived'].mean())

# Chipta narxlarini 4 ta teng guruhga bo'ladi.
# groupby orqali "Very High" (eng qimmat) chipta olganlarning omon qolish foizi "Low" (eng arzon) olganlardan qanchalik farq qilishini ko'rsatib beradi.
```

```python
df["age_group"]  = pd.qcut(df['Age'],q=3,labels=["child","adult","seinor"])
print(df.groupby("age_group")['Survived'].mean())

df["name_length"] =df['Name'].str.len()

# Ism uzunligi bo'yicha o'rtacha omon qolishni ko'rish
print(df.groupby(pd.cut(df['name_length'], bins=3))['Survived'].mean())
```

```python
# Yosh guruhlarini yaratish (cut orqali)
# 0-18 yosh: child, 18-50 yosh: adult, 50-100 yosh: seinor
bins = [0, 18, 50, 100]
labels = ["child", "adult", "seinor"]
df["age_group"] = pd.cut(df['Age'], bins=bins, labels=labels)

# 4. Natijalarni tahlil qilish
print("--- Yosh guruhlari bo'yicha omon qolish foizi ---")
print(df.groupby("age_group", observed=True)['Survived'].mean())
```

1. pd.qcut (Birinchi katakdagi natija)
Qanday ishlaydi: Datasetdagi odamlarni soni bo'yicha teng 3 qismga bo'ladi (har bir guruhda taxminan bir xil miqdorda odam bo'ladi).
Natijada chegaralar avtomatik ravishda yoshlar soniga qarab belgilanadi. Masalan:
child: ~0 dan 22 yoshgacha bo'lganlar (ko'pchiligi bolalar va o'smirlar).
adult: ~22 dan 38 yoshgacha bo'lganlar.
seinor: ~38 dan 80 yoshgacha bo'lganlar.
Bu yerda bolalar guruhiga yoshroq qatlam ko'proq kirib qolgani uchun foiz 0.41 (41%) chiqqan.
2. pd.cut (Ikkinchi katakdagi natija)
Qanday ishlaydi: Siz o'zingiz aniq chegaralarni berdingiz: bins = [0, 18, 50, 100].
child: 0 dan 18 yoshgacha bo'lganlar (aniq bolalar).
adult: 18 dan 50 yoshgacha bo'lganlar.
seinor: 50 dan 100 yoshgacha bo'lganlar.
Bu yerda haqiqiy yosh bolalar (0-18 yosh) soni kamroq bo'lishiga qaramay, ularning qutqarilish ehtimoli juda yuqori bo'lgani uchun foiz 0.50 (50% yoki 50.3%) gacha sakrab chiqdi.
Xulosa:
qcut — guruhlardagi odamlar sonini teng taqsimlaydi (statistik yondashuv).
cut — siz bergan aniq yosh chegaralari bo'yicha bo'ladi (mantiqiy yondashuv).

```python
# Label encoding
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

df["Sex_encoded"] = le.fit_transform(df["Sex"])

print(df[["Sex","Sex_encoded"]].head(4))
```

1. Label Encoding

Bu siz hozirgina ishlatgan usul. Har bir toifaga tartib bilan raqam beradi (0, 1, 2...).
Qanday ishlaydi:
female -> 0, male -> 1.
Qachon ishlatish kerak?
Agar toifalarda tartib (rank) bo'lsa (masalan: "kichik", "o'rta", "katta").
Agar toifalar soni juda ko'p bo'lsa.
Kamchiligi: Model adashib qolishi mumkin. Masalan, model "male" (1) > "female" (0) deb o'ylashi mumkin, aslida jinsda tartib yo'q.


2. One-Hot Encoding

Bu usul har bir toifa uchun alohida ustun yaratadi va u yerga 0 yoki 1 qo'yadi.
Qanday ishlaydi: Sex ustuni o'rniga Sex_female va Sex_male degan 2 ta ustun paydo bo'ladi.
Agar ayol bo'lsa: Sex_female = 1, Sex_male = 0.
Agar erkak bo'lsa: Sex_female = 0, Sex_male = 1.
Qachon ishlatish kerak?
Agar toifalarda tartib bo'lmasa (jins, shahar, port).
Toifalar soni kam bo'lsa.
Afzalligi: Model hech qanday noto'g'ri "tartib" (1 > 0) tushunchasini yaratmaydi.

```python
# Embarked
```

Embarked ustuni — bu yo'lovchi Titanik kemasiga qaysi portdan chiqqanini (kemaga o'tirgan joyini) bildiradi.

Datasetda u 3 ta harf bilan ifodalanadi:

C = Cherbourg (Fransiya)

Q = Queenstown (Irlandiya, hozirgi nomi Cobh)

S = Southampton (Angliya)


One-Hot Encoding —
asosan kategoriyalar soni kam bo'lgan ustunlar uchun ishlatiladi.
Masalan, Embarked (faqat 3 ta port bor: C, Q, S) yoki Sex (2 ta: male, female). Agar kategoriyalar ko'p bo'lib ketsa (masalan, 50 ta shahar),
One-Hot har biriga bittadan yangi ustun qo'shib, datasetni haddan tashqari kengaytirib yuboradi (bunga "curse of dimensionality" deyiladi).

Label Encoding —
kategoriyalar soni ko'p bo'lgan ustunlar uchun qulay. Masalan, yuzlab shaharlar yoki turli xil mahsulot nomlari bo'lsa, ularni 0, 1, 2, 3... deb bitta ustunning o'zida raqamlab chiqish qulayroq.

Lekin Label Encoding'da asosiy e'tibor kategoriyalarning tartibiga (ordinal) qaratiladi:
Agar kategoriyalar orasida mantiqiy tartib bo'lsa (masalan: Kichik = 0, O'rta = 1, Katta = 2), Label Encoding juda mos keladi.
Agar tartib bo'lmasa (masalan: S = 0, Q = 1, C = 2), model ularni "S kichikroq, C ulardan katta" deb noto'g'ri tushunib qolishi mumkin. Shuning uchun tartibsiz va kategoriyasi oz ustunlarga One-Hot Encoding afzal ko'riladi.

```python
# 1. One-Hot Encoding yaratish
embarked_dummies = pd.get_dummies(df["Embarked"], prefix="Embarked")

print("O'lchami:", embarked_dummies.shape)  # (889, 3) - 3 ta port
print(embarked_dummies.head(3))

# 2. Asosiy datasetga qo'shish (to'g'rilangan qism)
df = pd.concat([df, embarked_dummies], axis=1)

# Natijani tekshirish
print("\nYangi ustunlar bilan dataset o'lchami:", df.shape)
print(df[["Embarked", "Embarked_C", "Embarked_Q", "Embarked_S"]].head(3))
df
```

```python
df["is_alone"] = le.fit_transform(df["is_alone"])
df.head(2)
# LABEL ENCODING
```

```python
# ONE HOT ENCODING
title_dummies =  pd.get_dummies(df["Title"],prefix = "Title")

df = pd.concat([df, title_dummies], axis=1)
# Natijani tekshirish
print("\nYangi ustunlar bilan dataset o'lchami:", df.shape)

df
```

```python
# maxMin formulasi:x_scaled = (x-x_min)/ (x_max - x_min)
```

```python
from sklearn.preprocessing import MinMaxScaler


scaler_min_max = MinMaxScaler()
df["Age_scaled"]= scaler_min_max.fit_transform(df[["Age"]])
print(df[["Age","Age_scaled"]].head(3))
```

```python
# Std formulasi: x_scaled = (x-mean)/std
# Z-score


from sklearn.preprocessing import StandardScaler

scaler_std = StandardScaler()
df["Fare_std"]= scaler_std.fit_transform(df[["Fare"]])

print(f"mean: {scaler_std.mean_[0]:.2f}, std: {scaler_std.scale_[0]:.2f}")
print(df[["Fare","Fare_std"]].head(3))
print(df[["Fare","Fare_std"]].describe())
```

mean (o'rtacha) — bu hamma chipta narxlarining

o'rtacha qiymati: 32.10.
std (standart og'ish) — bu ma'lumotlar o'sha o'rtacha qiymatdan (32.10 dan)

o'rtacha hisobda qancha farq qilishini (o'ngga yoki chapga qancha og'ishini) ko'rsatuvchi o'lchov.

```python
# 1. MinMaxScaler uchun alohida obyekt
scaler_minmax = MinMaxScaler()
df["fare_per_person_mx"] = scaler_minmax.fit_temp = scaler_minmax.fit_transform(
    df[["fare_per_person"]]
)

# 2. StandardScaler uchun alohida obyekt
scaler_std = StandardScaler()
df["fare_per_person_std"] = scaler_std.fit_transform(df[["fare_per_person"]])

# Natijalarni tekshirish
print("--- MinMaxScaler natijasi ---")
print(df[["fare_per_person", "fare_per_person_mx"]].head(3))

print("\n--- StandardScaler natijasi ---")
print(df[["fare_per_person", "fare_per_person_std"]].head(3))

# Natijalarni tekshirish
print("--- MinMaxScaler  va Std natijasi Describe ---")
print(df[["fare_per_person_std", "fare_per_person_mx"]].describe())



```

Bugungi darsimizda Titanic dataseti ustida ishlashni davom ettirib, ma'lumotlarni mashina o'qitish modeliga tayyorlashning juda muhim bosqichlarini to'liq ko'rib chiqdik va amalda bajardik.

Bugun o'rgangan asosiy mavzu va tushunchalarimiz:

* Datasetni Yuklash va Dastlabki Tahlil: Kaggle'dan opendatasets yordamida Titanic ma'lumotlarini tortib oldik, df.shape, df.info() va df.isnull().sum() yordamida qator/ustunlar sonini hamda bo'sh (NaN) qiymatlarni aniqlab oldik.
* Ma'lumotlarni Tozalash (Data Cleaning): Keraksiz Cabin ustunini tashlab yubordik, Embarked ustunidagi bo'sh qatorlarni tozaladik va Age (yosh) ustunidagi bo'shliqlarni median qiymati bilan to'ldirdik.
* Feature Engineering (Yangi Ustunlar Yaratish):
* SibSp va Parch ustunlarini qo'shib, yo'lovchining oila a'zolari sonini bildiruvchi family_size hamda yolg'iz kelganini aniqlovchi is_alone ustunlarini yasadik.
* Regular Expression (`Regex`) yordamida Name ustunidan odamlarning unvonlarini (`Title`: Mr, Miss, Mrs, Master va hokazo) sug'urib oldik va kam uchraydiganlarini "Rare" guruhiga birlashtirdik.
* Har bir kishiga to'g'ri keladigan chipta narxini hisoblab chiqarish uchun fare_per_person ustunini yaratdik.


* Guruhlash va Tahlil (`cut` va `qcut`):
* pd.cut yordamida yoshlarni aniq oraliqlarga (`child`, adult, `seinor`) bo'ldik.
* pd.qcut yordamida esa ma'lumotlarni soni bo'yicha teng qismlarga taqsimlab, turli guruhlarning omon qolish (`Survived`) foizlarini taqqoslab ko'rdik.


* Encoding (Matnni Songa Aylantirish):
* Label Encoding: Sex va is_alone kabi ustunlarni 0 va 1 raqamlariga o'tkazdik (`LabelEncoder`).
* One-Hot Encoding: Embarked va Title kabi kategorik ustunlar uchun pd.get_dummies yordamida alohida binar (True/False yoki 1/0) ustunlar yaratib, datasetni kengaytirdik.


* Feature Scaling (Masshtablash):
* MinMaxScaler: Qiymatlarni 0 va 1 oralig'iga keltiruvchi $x_{scaled} = \frac{x - x_{min}}{x_{max} - x_{min}}$ formulasini qo'lladik.
* StandardScaler: O'rtacha qiymati 0, standart og'ishi 1 bo'ladigan Z-score ($x_{scaled} = \frac{x - mean}{std}$) usulini o'rganib, Age va Fare kabi ustunlarni standartlashtirdik hamda ularning mean va std ko'rsatkichlarini tahlil qildik.
