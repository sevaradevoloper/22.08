turnaround_reason — Cho'qqiga chiqmay ortga qaytish sababi (masalan, exhaustion — charchash, altitude sickness — tog' kasalligi, team_decision — jamoa qarori va h.k.).

```python
# Bizada juda ham kop ustunlar bor ekan shunga kk ajratib olamiz
# Asl df o'zgarishsiz qoladi, biz faqat model uchun alohida o'zgaruvchi olamiz
X = df[
    [
        "age",
        "years_climbing",
        "prior_8000m_peaks",
        "o2_start_altitude_m",
        "queue_minutes_above_8000",
        "weather_score",
    ]
]
y = df["summited"]
```

```python
X.describe()

Q1 = X.quantile(0.25)
Q3 = X.quantile(0.75)
IQR = Q3 - Q1

pastki = Q1 - 1.5 * IQR
yuqori = Q3 + 1.5 * IQR
outlierlar = X <pastki
print(f"Outlierlar soni: {len(outlierlar)}")
```

```python
for col in X.columns:
  mean = X[col].mean()
  std = X[col].std()
  z_scores = (X[col] - mean) / std
  outliers_count = (z_scores.abs() > 3).sum()
  print(f"{col}: {outliers_count} ta outlier")
```

```python
# Masalan, IQR bo'yicha chegaralarni topib:
# pastki va yuqori chegaralar bo'yicha X ni tozalash:
X_clean = X[
    (X >= pastki) & (X <= yuqori)
].dropna()  # yoki faqat bitta ustun uchun tekshirishingiz mumkin

# Qancha ma'lumot o'chib ketganini tekshirish:
print("Eski X qatorlar soni:", X.shape[0])
print("Yangi tozalangan X qatorlar soni:", X_clean.shape[0])
print("O'chirilgan outlierlar soni:", X.shape[0] - X_clean.shape[0])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1 ta figura va 2 ta ustun (grafik) ochamiz
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# 1-chi grafik (masalan, age taqsimoti)
sns.histplot(X_clean["age"], ax=axes[0], color="blue")
axes[0].set_title("Age Distribution")

# 2-chi grafik (masalan, weather_score taqsimoti)
sns.histplot(X_clean["weather_score"], ax=axes[1], color="green")
axes[1].set_title("Weather Score Distribution")

# Chiroyli ko'rinishi uchun joylashuvni to'g'rilaymiz
plt.tight_layout()
plt.show()
```

```python
# 1 ta figura va 2 ta ustun (grafik) ochamiz
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.heatmap(X, annot=True, cmap='coolwarm', fmt='.2f',ax=axes[0])
axes[0].set_title("Oldin")

sns.heatmap(X_clean, annot=True, cmap='coolwarm', fmt='.2f',ax=axes[1])
axes[1].set_title("Keyin")

# Chiroyli ko'rinishi uchun joylashuvni to'g'rilaymiz
plt.tight_layout()
plt.show()
```
