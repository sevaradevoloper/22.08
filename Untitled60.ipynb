```python
import numpy as np
import matplotlib.pyplot as plt




#21 22 23 mavzular
# 1. Funksiyani aniqlaymiz: x=2 nuqtada 0/0 bo'ladi (teshik)
def f(x):
    return (x**2 - 4) / (x - 2)

# 2. X qiymatlarini tayyorlaymiz (2 nuqtani o'tkazib yuboramiz)
x = np.linspace(0, 4, 200)
x = x[x != 2]
y = f(x)

# 3. Grafik chizamiz
plt.figure(figsize=(8, 5))
plt.plot(x, y, label="f(x) = (x² - 4) / (x - 2)", color="blue", linewidth=2)

# X=2 nuqtadagi teshikni (limit=4) qizil aylana bilan belgilaymiz
plt.scatter([2], [4], color="red", s=100, facecolors='none', edgecolors='red', linewidth=2, label="Teshik nuqta (Limit = 4)")

plt.title("Limit va Uzluksizlik: x=2 da funksiya yo'q, lekin Limit = 4 bor!")
plt.xlabel("x qiymati")
plt.ylabel("f(x) qiymati")
plt.axvline(x=2, color='gray', linestyle='--', alpha=0.5)
plt.axhline(y=4, color='gray', linestyle='--', alpha=0.5)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Kosa shaklidagi funksiya: f(x) = x^2
def f(x):
    return x**2

# 2. Hosila (qiyalik) formulasi: f'(x) = 2x
def df(x):
    return 2*x

x = np.linspace(-4, 4, 100)
y = f(x)

# X=3 nuqtadagi urinma (qiyalik)
x_nuqta = 3
y_nuqta = f(x_nuqta)
qiyalik = df(x_nuqta) # 2*3 = 6

# Urinma chizig'i tenglamasi
urinma_x = np.linspace(1, 5, 10)
urinma_y = qiyalik * (urinma_x - x_nuqta) + y_nuqta

plt.figure(figsize=(8, 5))
plt.plot(x, y, label="Cost Funksiya: f(x) = x²", color="black")
plt.plot(urinma_x, urinma_y, label=f"x=3 dagi Hosila (Qiyalik) = {qiyalik}", color="red", linestyle="--")
plt.scatter([x_nuqta], [y_nuqta], color="red", s=80)

plt.title("Hosila — Grafikning Berilgan Nuqtadagi Qiyaligi")
plt.xlabel("x (Model parametri)")
plt.ylabel("f(x) (Xato miqdori)")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Ma'lumotlar (y = 2x bo'lishi kerak)
x_data = np.array([1, 2, 3, 4])
y_data = np.array([2, 4, 6, 8])

# 2. Cost function (MSE)
def cost(w):
    pred = w * x_data
    return np.mean((pred - y_data) ** 2)

# 3. Gradient
def gradient(w):
    pred = w * x_data
    return np.mean(2 * x_data * (pred - y_data))

# 4. Gradient Descent jarayoni
w = 0.0          # Boshlang'ich taxmin
lr = 0.05        # Learning rate (Qadam kattaligi)
history_w = [w]
history_cost = [cost(w)]

for step in range(10):
    g = gradient(w)
    w = w - lr * g
    history_w.append(w)
    history_cost.append(cost(w))

# 5. Grafik chizish
w_vals = np.linspace(-0.5, 3.5, 100)
cost_vals = [cost(w_i) for w_i in w_vals]

plt.figure(figsize=(9, 5))
plt.plot(w_vals, cost_vals, label="Cost Funksiya Parabolasi C(w)", color="gray", linestyle="--")
plt.plot(history_w, history_cost, 'ro-', label="Gradient Descent Qadamlari (w=0 dan w=2 ga)", markersize=6)

plt.title("Gradient Descent: Model Xatoni Kamaytirib w=2 Minimumga Yetib Bormoqda")
plt.xlabel("Parametr w")
plt.ylabel("Xato C(w)")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
