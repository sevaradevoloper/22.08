```python
Q1 = df["release_year"].quantile(0.25)
Q3 = df["release_year"].quantile(0.75)
IQR = Q3 - Q1
print(IQR)

pastki = Q1 - 1.5 * IQR
yuqori = Q3 + 1.5 * IQR
outlierlar = df[df["release_year"]<pastki]
print(f"Outlierlar soni: {len(outlierlar)}")
```

```python
mean = df["release_year"].mean()
std = df["release_year"].std()

df ['z_score'] = (df["release_year"]-mean)/std
print((df["z_score"].abs()>3).sum())
```

```python
# 5. Uyga vazifa (Oson): 'listed_in' dagi NaN soni va 'rating' taqsimotini tekshiramiz
print("NaN listed_in:", df["listed_in"].isnull().sum())
print(df["rating"].value_counts())

# 6. Uyga vazifa (O'rta): 'duration' matnidan faqat sonni ajratib yangi ustunga o'tkazamiz
df["duration_num"] = df["duration"].str.extract("(\d+)").astype(float)


# 7. Uyga vazifa (Qiyin): Barcha tozalash adamlari jamlangan to'liq pipeline funksiyasi
def clean_netflix_data(df):
    df = df.dropna(subset=["date_added", "rating", "duration"]).fillna(
        {"director": "Noma'lum", "cast": "Noma'lum", "country": "Noma'lum"}
    )
    df = df.drop_duplicates()
    df["is_classic"] = df["release_year"] < 1999
    df["duration_num"] = df["duration"].str.extract("(\d+)").astype(float)
    return df


# 8. Yakuniy natijani tekshirish
df_clean = clean_netflix_data(pd.read_csv("netflix_titles.csv"))
print("Tozalangan dataset o'lchami:", df_clean.shape)
```
