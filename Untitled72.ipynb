```python
print(df["release_year"].describe())
```

```python
Q1 = df["release_year"].quantile(0.25)
Q3 = df["release_year"].quantile(0.75)
IQR = Q3 - Q1
print(IQR)

pastki = Q1 - 1.5 * IQR
yuqori = Q3 + 1.5 * IQR
outlierlar = df[df["release_year"]<pastki]
print(f"Outlierlar soni: {len(outlierlar)}")
```

```python
mean = df["release_year"].mean()
std = df["release_year"].std()

df ['z_score'] = (df["release_year"]-mean)/std
print((df["z_score"].abs()>3).sum())
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Mean va std ni shu joyning o'zida qaytadan hisoblaymiz
mean = df["release_year"].mean()
std = df["release_year"].std()

# 2. Grafikni chizamiz
plt.figure(figsize=(10, 5))
plt.hist(
    df["release_year"],
    bins=20,
    color="#2ECB9B",
    alpha=0.7,
    edgecolor="white",
)
plt.axvline(mean, color="red", linestyle="--", lw=2, label=f"Mean={mean:.1f}")
plt.axvline(
    mean + std, color="orange", linestyle=":", lw=2, label=f"+1σ={mean+std:.1f}"
)
plt.axvline(
    mean - std, color="orange", linestyle=":", lw=2, label=f"-1σ={mean-std:.1f}"
)

plt.title("Release Year taqsimoti")
plt.xlabel("Yil")
plt.ylabel("Soni")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# 1. IQR usuli bo'yicha outlier sharti
iqr_outliers_mask = (df["release_year"] < (Q1 - 1.5 * IQR)) | (
    df["release_year"] > (Q3 + 1.5 * IQR)
)

# 2. Z-score usuli bo'yicha outlier sharti
z_outliers_mask = df["z_score"].abs() > 3

# Endi sonlarini solishtiramiz
print(f"IQR usuli bo'yicha outlier'lar soni: {iqr_outliers_mask.sum()}")
print(f"Z-score usuli bo'yicha outlier'lar soni: {z_outliers_mask.sum()}")
```

```python
print(
    f"IQR ko'p: {iqr_outliers_mask.sum()}"
    if iqr_outliers_mask.sum() > z_outliers_mask.sum()
    else f"Z-score ko'p: {z_outliers_mask.sum()}"
)
```

```python
if iqr_outliers_mask.sum() > z_outliers_mask.sum():
  print(f"IQR ko'p: {iqr_outliers_mask.sum()}")
else:
  print(f"Z-score ko'p: {z_outliers_mask.sum()}")
```
