21. NumPy yordamida ikkita vektorni qabul qilib, ularning yig'indisini, ayirmasini va skalyar (dot) ko'paytmasini hisoblovchi funksiya yozing.  (6 ball)


```python
import numpy as np

v1 = np.array([1,2,3])
v2 = np.array([4,5,6])
q = v1+v2
a = v2 - v1
dot = v1 @ v2
print(f'vektorlarni qoshish: {q}')
print(f'vektorlarni ayrish: {a}')
print(f'vektorlarni kopaytirish: {dot}')
```

22. NumPy yordamida berilgan 2x2 yoki 3x3 matritsaning determinanti va inverse'ini hisoblovchi funksiya yozing (np.linalg.det, np.linalg.inv). Agar inverse mavjud bo'lmasa, mos xabar chiqaring.  (6 ball)


```python
m = np.array([[1,2],
               [3,4]
               ])
det = np.linalg.det(m)
#qolda :  1*4 -2 *3
print(det)

m2= np.array([[1,2],
               [4,10]
               ])
det = np.linalg.det(m2)
#qolda :  1*10 -2 * 4
print(det)

inv = np.linalg.inv(m)
print(f'inversedan oldin{m}')
#inv formulasi = > det chiqandan keyin 1/det(A) * teskari matrix
print(f'inverse qiymat {inv}')
```

24. f(x) = x^2 funksiyasi uchun sonli (numerical) hosilani finite difference usuli bilan hisoblovchi funksiya yozing: f'(x) ≈ (f(x+h) - f(x-h)) / (2h). x=3 nuqtada natijani chop eting va aniq hosila (2x=6) bilan solishtiring.  (6 ball)


```python
# Finite difference usuli bilan sonli hosilani hisoblovchi funksiya ,h=1e-5 bu juda kichik son, masalan 0.00001. Hosila aniq chiqishi uchun h ni shunaqa mitti qilib olamiz
def numerical_derivative(f, x, h=1e-5):
  # Markaziy ayirma formulasi (Central Difference)
  return (f(x + h) - f(x - h)) / (2 * h)


# Berilgan funksiya: f(x) = x^2
def f(x):
  return x**2


# Nuqta va qadam kattaligi
x_val = 3
h_val = 1e-5

# Sonli hosilani topamiz
num_deriv = numerical_derivative(f, x_val, h_val)

# Aniq (analitik) hosila: 2*x
exact_deriv = 2 * x_val

# Natijalarni chop etamiz
print(f"Sonli hosila : {num_deriv:.5f}")
print(f"Aniq hosila : {exact_deriv}")
print(f"Farqi : {(num_deriv - exact_deriv)}")
# hulosa qasyi x soni 2 ga kopaytirsa 6 chiqadi bu bizada 3 shuni tekshirdik yani 2* 3 qildik = 6 chiqdi funkisiya togri ishladi
```

25. Cost function C(w) = (w - 3)^2 uchun oddiy gradient descent algoritmini yozing: boshlang'ich w=0, learning_rate=0.1, 20 iteratsiya. Har ikki qadamda w qiymatini chop eting va w qanday 3 ga yaqinlashishini kuzating.  (6 ball)


```python
# w va lr yozamiz
w = 0.0
lr = 0.1

print("Iteratsiya | w qiymati")
print("-" * 25)
#tsikl ishlatamiz
for i in range(1, 21):
  # Hosilani topamiz: C(w) = (w - 3)^2 ning hosilasi = 2 * (w - 3)
  hosila = 2 * (w - 3)

  # Gradient descent formulasi bo'yicha w ni yangilaymiz
  w = w - lr * hosila

  # Shart bo'yicha: har ikki qadamda (2, 4, 6, ...) chop etamiz
  if i % 2 == 0:
    print(f"  {i:2d}       | {w:.4f}")
```
