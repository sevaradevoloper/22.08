```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Dastlabki ma'lumotni olamiz (masalan, Titanic Fare ustuni)
# df['Fare'] da outlier'lar borligini bilamiz
fare_data = df['Fare'].dropna()

# 2. IQR usuli orqali chegaralarni topamiz (Q1, Q3 va IQR)
Q1 = fare_data.quantile(0.25)
Q3 = fare_data.quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Pastki chegara: {lower_bound}, Yuqori chegara: {upper_bound}")

# 3. Capping (Chegaralash) usuli bilan outlier'larni to'ldiramiz / moslaymiz
# Yuqori chegaradan kattalarini yuqori chegaraning o'ziga tenglaymiz (yoki kesib tashlaymiz)
df_cleaned = df.copy()
df_cleaned['Fare_cleaned'] = np.where(
    df_cleaned['Fare'] > upper_bound, upper_bound,
    np.where(df_cleaned['Fare'] < lower_bound, lower_bound, df_cleaned['Fare'])
)

# --- 4. SOLISHTIRISH (DO va POSLE holati grafikda) ---
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# TOZALASHDAN OLDIN (Before / Do)
axes[0].boxplot(df['Fare'].dropna())
axes[0].set_title('DO (Outlier\'lar bilan): Fare')
axes[0].set_ylabel('Narx')

# TOZALASHDAN KEYIN (After / Posle)
axes[1].boxplot(df_cleaned['Fare_cleaned'])
axes[1].set_title('POSLE (Outlier\'lar tozalan / chegaralangandan keyin): Fare')
axes[1].set_ylabel('Narx')

plt.tight_layout()
plt.show()
```
