```python
%%manim -qm -v WARNING LimitAndContinuityMasterclass

from manim import *
from gtts import gTTS
import os

def generate_voiceover(text, filename):
    """Matnni mp3 audio faylga o'girish (lang='tr' o'zbekcha matnga eng yaqin talaffuz beradi)"""
    tts = gTTS(text=text, lang='tr')
    tts.save(filename)

class LimitAndContinuityMasterclass(Scene):
    def construct(self):
        # ---------------------------------------------------------
        # 1. AUDIO FAYLLARNI YARATISH (Har bir bosqich uchun)
        # ---------------------------------------------------------
        text1 = "Birinchi bosqich. Limit bu funksiyaning ma'lum bir nuqtaga yaqinlashgandagi niyatidir. Nuqtaning o'zida teshik bo'lsa ham, limit mavjud bo'la oladi."
        text2 = "Ikkinchi bosqich. Funksiya uzluksiz bo'lishi uchun unda teshik, sakrash va portlash uzilishlari bo'lmasligi kerak."
        text3 = "Uchinchi bosqich. Gradient Descent algoritmi model xatoligini kamaytirish uchun tepalikdan pastga tushadi. Funksiya uzluksiz bo'lsa, gradient to'g'ri hisoblanadi."

        generate_voiceover(text1, "voice_part1.mp3")
        generate_voiceover(text2, "voice_part2.mp3")
        generate_voiceover(text3, "voice_part3.mp3")

        # =========================================================
        # 1-BOSQICH: LIMIT VA TESHIK NUQTA (0/0)
        # =========================================================
        self.add_sound("voice_part1.mp3")

        title1 = Text("1-Bosqich: Limit Tushunchasi va Teshik Nuqta", font_size=24, color=YELLOW).to_edge(UP)

        axes1 = Axes(
            x_range=[0, 4, 1], y_range=[0, 5, 1],
            x_length=4.5, y_length=3.8,
            axis_config={"color": GREY_B}
        ).to_corner(DL).shift(UP * 0.2 + RIGHT * 0.2)
        axes1_labels = axes1.get_axis_labels(x_label="x", y_label="y")

        curve1 = axes1.plot(lambda x: x + 1, color=BLUE, x_range=[0.1, 3.5])
        hole_circle = Circle(radius=0.09, color=RED, fill_color=BLACK, fill_opacity=1).move_to(axes1.c2p(1, 2))

        x_tracker = ValueTracker(0.3)
        dot_left = always_redraw(lambda: Dot(
            axes1.c2p(1 - x_tracker.get_value(), 2 - x_tracker.get_value()),
            color=GREEN, radius=0.08
        ))
        dot_right = always_redraw(lambda: Dot(
            axes1.c2p(1 + x_tracker.get_value(), 2 + x_tracker.get_value()),
            color=PURPLE, radius=0.08
        ))

        limit_formula = always_redraw(lambda: MathTex(
            f"\\lim_{{x \\to 1}} \\frac{{x^2 - 1}}{{x - 1}} = 2 \\quad (f(1) \\text{{ MAVJUD EMAS}})",
            font_size=18, color=YELLOW
        ).next_to(title1, DOWN, buff=0.15))

        konspekt_box1 = Rectangle(width=6.2, height=4.8, color=BLUE_A, fill_opacity=0.15).to_corner(DR).shift(UP * 0.1)
        konspekt_text1 = Text(
            "KONSPEKT (1-Qism):\n\n"
            "• Limit — x->a ga yaqinlashganda f(x) ning\n"
            "  'niyati' (qayerga borishi).\n\n"
            "• f(a) — funksiyaning x=a nuqtadagi HAQIQIY qiymati.\n\n"
            "• Teshik Nuqta: f(a) mavjud bo'lmasligi (0/0) mumkin,\n"
            "  lekin limit baribir MAVJUD bo'ladi!",
            font_size=13, line_spacing=1.2
        ).move_to(konspekt_box1.get_center())

        self.play(Write(title1), Create(axes1), Write(axes1_labels), Create(curve1), Create(hole_circle))
        self.play(FadeIn(dot_left), FadeIn(dot_right), Write(limit_formula), Create(konspekt_box1), Write(konspekt_text1))

        # Nuqtalar yaqinlashishi va audio tugashini kutish
        self.play(x_tracker.animate.set_value(0.01), run_time=5.0, rate_func=smooth)
        self.wait(5)  # 1-audio to'liq tugashi uchun kutiladi

        self.play(*[Unwrite(m) for m in [title1, axes1, axes1_labels, curve1, hole_circle, dot_left, dot_right, limit_formula, konspekt_box1, konspekt_text1]])
        self.wait(1)

        # =========================================================
        # 2-BOSQICH: UZLUKSIZLIK VA UZILISH TURLARI
        # =========================================================
        self.add_sound("voice_part2.mp3")

        title2 = Text("2-Bosqich: Uzluksizlik va Uzilish Turlari", font_size=24, color=TEAL).to_edge(UP)

        axes2 = Axes(
            x_range=[-2, 3, 1], y_range=[-1, 4, 1],
            x_length=4.5, y_length=3.8,
            axis_config={"color": GREY_B}
        ).to_corner(DL).shift(UP * 0.2 + RIGHT * 0.2)

        curve2_left = axes2.plot(lambda x: 1, x_range=[-1.8, 0], color=GREEN)
        curve2_right = axes2.plot(lambda x: 2.5, x_range=[0, 2.5], color=PURPLE)
        dot_jump1 = Dot(axes2.c2p(0, 1), color=GREEN, radius=0.08)
        dot_jump2 = Dot(axes2.c2p(0, 2.5), color=PURPLE, radius=0.08)

        limit_spec = MathTex(
            "\\lim_{{x \\to 0^-}} f(x) = 1 \\neq \\lim_{{x \\to 0^+}} f(x) = 2.5 \\implies \\text{Uzilish}",
            font_size=18, color=RED
        ).next_to(title2, DOWN, buff=0.15)

        konspekt_box2 = Rectangle(width=6.2, height=4.8, color=TEAL, fill_opacity=0.15).to_corner(DR).shift(UP * 0.1)
        konspekt_text2 = Text(
            "KONSPEKT (Uzluksizlik va Uzilishlar):\n\n"
            "• 3 ta shart: 1) f(a) bor  2) Limit bor  3) Limit = f(a)\n\n"
            "• Uzilish turlari:\n"
            "  1. Teshik (Removable): Limit bor, f(a) yo'q/teng emas.\n"
            "  2. Sakrash (Jump): Chap limit != O'ng limit.\n"
            "  3. Portlash (Infinite): Funksiya +-inf ga qochadi (1/x).",
            font_size=12, line_spacing=1.2
        ).move_to(konspekt_box2.get_center())

        self.play(Write(title2), Create(axes2), Create(curve2_left), Create(curve2_right), FadeIn(dot_jump1), FadeIn(dot_jump2))
        self.play(Write(limit_spec), Create(konspekt_box2), Write(konspekt_text2))

        # 2-audio to'liq eshitilishi uchun kutish
        self.wait(8)

        self.play(*[Unwrite(m) for m in [title2, axes2, curve2_left, curve2_right, dot_jump1, dot_jump2, limit_spec, konspekt_box2, konspekt_text2]])
        self.wait(1)

        # =========================================================
        # 3-BOSQICH: GRADIENT DESCENT DINAMIK VISUALISATION
        # =========================================================
        self.add_sound("voice_part3.mp3")

        title3 = Text("3-Bosqich: Gradient Descent qanday ishlaydi?", font_size=24, color=GOLD).to_edge(UP)

        axes3 = Axes(
            x_range=[-2, 2, 1], y_range=[0, 4, 1],
            x_length=4.5, y_length=3.8,
            axis_config={"color": GREY_B}
        ).to_corner(DL).shift(UP * 0.2 + RIGHT * 0.2)
        axes3_labels = axes3.get_axis_labels(x_label="w (vazn)", y_label="J(w) (xato)")

        loss_curve = axes3.plot(lambda w: w**2, color=ORANGE, x_range=[-1.8, 1.8])

        w_tracker = ValueTracker(1.6)

        gd_dot = always_redraw(lambda: Dot(
            axes3.c2p(w_tracker.get_value(), w_tracker.get_value()**2),
            color=RED, radius=0.12
        ))

        def get_tangent_line():
            w = w_tracker.get_value()
            y = w**2
            slope = 2 * w
            dx = 0.6
            p1 = axes3.c2p(w - dx, y - slope * dx)
            p2 = axes3.c2p(w + dx, y + slope * dx)
            return Line(p1, p2, color=YELLOW, stroke_width=3)

        tangent_line = always_redraw(get_tangent_line)

        konspekt_box3 = Rectangle(width=6.2, height=4.8, color=GOLD_B, fill_opacity=0.15).to_corner(DR).shift(UP * 0.1)
        konspekt_text3 = Text(
            "KONSPEKT (Gradient Descent):\n\n"
            "• Gradient = Funksiyaning hosilasi (qiyalik).\n\n"
            "• Maqsad: Xatolik (J) ni minimal nuqtaga tushirish.\n\n"
            "• Uzluksizlik roli: Funksiya silliq va uzluksiz bo'lsa,\n"
            "  gradient har bir nuqtada hisoblanadi va to'p\n"
            "  tubiga muvaffaqiyatli yetib boradi!",
            font_size=12, line_spacing=1.2
        ).move_to(konspekt_box3.get_center())

        self.play(Write(title3), Create(axes3), Write(axes3_labels), Create(loss_curve))
        self.play(FadeIn(gd_dot), Create(tangent_line), Create(konspekt_box3), Write(konspekt_text3))

        # Ballning pastga harakatlanishi va 3-audioning tugashi
        self.play(w_tracker.animate.set_value(0.0), run_time=6.0, rate_func=smooth)
        self.wait(4)
```

```python
# ==============================================================================
# NAJOT TA'LIM - MACHINE LEARNING KURSI
# DARS 21: Limit va uzluksizlik (Manim Video-Darslik Kodlari)
# ==============================================================================
# Colab-ning birinchi yacheykasida quyidagi buyruqni ishga tushirishni unutmang:
# !pip install manim gtts pydub moviepy
# !apt-get install ffmpeg
# ==============================================================================

# ==============================================================================
# CELL 1: MAVZU 1 - LIMIT TUSHUNCHASI (0/0 HOALTI VA YAQINLASHISH)
# ==============================================================================
%%manim -qm -v WARNING LimitTushunchasi

from manim import *
from gtts import gTTS
from pydub import AudioSegment
from moviepy.editor import VideoFileClip, AudioFileClip
import os

def generate_voiceover(text, filename):
    tts = gTTS(text=text, lang='tr')
    tts.save(filename)

class LimitTushunchasi(Scene):
    def construct(self):
        # 1. Ovoz yaratish
        voice_text = (
            "Limit tushunchasi. Funksiya x teng ikkida nol bo'lingan nol aniqlanmagan holatga keladi. "
            "Lekin ikkiga yaqinlashganimizda f x qiymati to'rtga intiladi. "
            "Limit funksiyaning shu nuqtadagi emas, atrofdagi harakatini ko'rsatadi."
        )
        generate_voiceover(voice_text, "voice_cell1.mp3")

        # Title
        main_title = Text("1-Mavzu: Limit Tushunchasi (0/0 Holati)", font_size=24, color=BLUE_A).to_edge(UP, buff=0.3)
        self.add(main_title)

        # CHAP TOMON: Grafika (Axes va Funksiya)
        axes = Axes(
            x_range=[-0.5, 4.5, 1],
            y_range=[-0.5, 6.5, 1],
            x_length=5.5,
            y_length=4.5,
            axis_config={"include_numbers": True, "font_size": 16}
        ).to_corner(DL, buff=0.6)

        # f(x) = (x^2 - 4) / (x - 2) -> x + 2 (x != 2)
        graph = axes.plot(lambda x: x + 2, x_range=[-0.5, 3.8], color=TEAL, stroke_width=4)
        graph_label = MathTex("f(x) = \\frac{x^2 - 4}{x - 2}", font_size=28, color=TEAL).next_to(axes, UP, buff=0.2)

        # Hole at (2, 4)
        hole = Dot(axes.c2p(2, 4), radius=0.09, color=RED, fill_opacity=0, stroke_width=2.5)

        # Tracker for moving point
        x_tracker = ValueTracker(0.5)

        dot = always_redraw(lambda: Dot(
            axes.c2p(x_tracker.get_value(), x_tracker.get_value() + 2),
            color=YELLOW,
            radius=0.1
        ))

        v_line = always_redraw(lambda: axes.get_vertical_line(
            axes.c2p(x_tracker.get_value(), x_tracker.get_value() + 2),
            color=GRAY,
            line_func=DashedLine
        ))

        # Value indicators
        val_tex = always_redraw(lambda: MathTex(
            f"x = {x_tracker.get_value():.3f}, \\quad f(x) = {x_tracker.get_value() + 2:.3f}",
            font_size=22, color=YELLOW
        ).next_to(axes, DOWN, buff=0.3))

        left_group = VGroup(axes, graph, graph_label, hole, dot, v_line, val_tex)

        # O'NG TOMON: Konspekt qutisi (Rectangle)
        box = Rectangle(width=5.8, height=5.2, color=BLUE, fill_opacity=0.15).to_corner(DR, buff=0.5)
        box_title = Text("KONSPEKT: LIMIT", font_size=20, weight=BOLD, color=YELLOW).next_to(box.get_top(), DOWN, buff=0.3)

        line1 = MathTex("\\lim_{x \\to a} f(x) = L", font_size=26, color=WHITE)
        line2 = Text("1. x = 2 da f(2) = 0/0 (Aniqlanmagan!)", font_size=16, color=RED_B)
        line3 = MathTex("\\text{Algebra: } \\frac{(x-2)(x+2)}{x-2} = x + 2", font_size=22, color=GREEN_B)
        line4 = Text("2. x -> 2 da f(x) -> 4 (Limit bor!)", font_size=16, color=GREEN_A)
        line5 = Text("3. Limit f(a) ning o'ziga bog'liq emas,", font_size=15, color=LIGHT_GRAY)
        line6 = Text("   atrofidagi harakatga bog'liq.", font_size=15, color=LIGHT_GRAY)

        notes = VGroup(line1, line2, line3, line4, line5, line6).arrange(DOWN, aligned_edge=LEFT, buff=0.25).next_to(box_title, DOWN, buff=0.3)
        right_group = VGroup(box, box_title, notes)

        # Animatsiyalar
        self.play(Create(axes), Create(box), Write(box_title))
        self.play(Create(graph), Write(graph_label), Create(hole))
        self.play(Write(notes))
        self.play(Create(dot), Create(v_line), Write(val_tex))

        # Chapdan yaqinlashish (x: 0.5 -> 1.999)
        self.play(x_tracker.animate.set_value(1.999), run_time=3.5, rate_func=linear)
        self.wait(0.5)

        # O'ngdan yaqinlashish (x: 3.5 -> 2.001)
        self.play(x_tracker.animate.set_value(3.5), run_time=0.8)
        self.play(x_tracker.animate.set_value(2.001), run_time=3.5, rate_func=linear)
        self.wait(2)

    def on_finish(self):
        super().on_finish(self)
        if os.path.exists("voice_cell1.mp3"):
            a1 = AudioSegment.from_mp3("voice_cell1.mp3")
            a1.export("full_voice1.mp3", format="mp3")

            video_path = self.renderer.file_writer.movie_file_path
            video_clip = VideoFileClip(video_path)
            audio_clip = AudioFileClip("full_voice1.mp3")

            final_clip = video_clip.set_audio(audio_clip)
            final_clip.write_videofile("final_output_LimitTushunchasi.mp4", codec="libx264", audio_codec="aac")
```

```python
# ==============================================================================
# CELL 2: MAVZU 2 - BIR TOMONLAMA LIMITLAR VA CHEKSIZLIK
# ==============================================================================
%%manim -qm -v WARNING BirTomonlamaLimitlar

from manim import *
from gtts import gTTS
from pydub import AudioSegment
from moviepy.editor import VideoFileClip, AudioFileClip
import os

def generate_voiceover(text, filename):
    tts = gTTS(text=text, lang='tr')
    tts.save(filename)

class BirTomonlamaLimitlar(Scene):
    def construct(self):
        # 1. Ovoz yaratish
        voice_text = (
            "Bir tomonlama limitlar. Chapdan kelganda limit minus bir, o'ngdan kelganda plyus bir. "
            "Chap va o'ng limitlar teng bo'lmagani uchun umumiy limit mavjud emas."
        )
        generate_voiceover(voice_text, "voice_cell2.mp3")

        # Title
        main_title = Text("2-Mavzu: Bir Tomonlama Limitlar & Sakrash", font_size=24, color=PURPLE_A).to_edge(UP, buff=0.3)
        self.add(main_title)

        # CHAP TOMON: Grafika f(x) = |x| / x
        axes = Axes(
            x_range=[-3.5, 3.5, 1],
            y_range=[-2.5, 2.5, 1],
            x_length=5.5,
            y_length=4.5,
            axis_config={"include_numbers": True, "font_size": 16}
        ).to_corner(DL, buff=0.6)

        left_branch = axes.plot(lambda x: -1, x_range=[-3.2, -0.05], color=RED, stroke_width=4)
        right_branch = axes.plot(lambda x: 1, x_range=[0.05, 3.2], color=GREEN, stroke_width=4)

        hole_left = Dot(axes.c2p(0, -1), radius=0.08, color=RED, fill_opacity=0, stroke_width=2)
        hole_right = Dot(axes.c2p(0, 1), radius=0.08, color=GREEN, fill_opacity=0, stroke_width=2)

        graph_label = MathTex("f(x) = \\frac{|x|}{x}", font_size=28, color=PURPLE_B).next_to(axes, UP, buff=0.2)

        left_dot = Dot(axes.c2p(-3, -1), color=YELLOW, radius=0.1)
        right_dot = Dot(axes.c2p(3, 1), color=ORANGE, radius=0.1)

        # O'NG TOMON: Konspekt qutisi
        box = Rectangle(width=5.8, height=5.2, color=PURPLE, fill_opacity=0.15).to_corner(DR, buff=0.5)
        box_title = Text("KONSPEKT: BIR TOMONLAMA", font_size=20, weight=BOLD, color=YELLOW).next_to(box.get_top(), DOWN, buff=0.3)

        line1 = MathTex("\\text{Chap limit: } \\lim_{x \\to 0^-} f(x) = -1", font_size=22, color=RED_B)
        line2 = MathTex("\\text{O'ng limit: } \\lim_{x \\to 0^+} f(x) = +1", font_size=22, color=GREEN_B)
        line3 = Text("QOIDA: Chap limit != O'ng limit", font_size=16, color=ORANGE)
        line4 = Text("=> lim (x->0) f(x) MAVJUD EMAS!", font_size=16, color=RED)
        line5 = Text("ML DA UCHRAYDI:", font_size=16, color=YELLOW)
        line6 = Text("ReLU g(x)=max(0,x) x=0 da kelishgan,", font_size=14, color=WHITE)
        line7 = Text("lekin burchak bor va uzluksiz!", font_size=14, color=WHITE)

        notes = VGroup(line1, line2, line3, line4, line5, line6, line7).arrange(DOWN, aligned_edge=LEFT, buff=0.2).next_to(box_title, DOWN, buff=0.25)

        # Animatsiyalar
        self.play(Create(axes), Create(box), Write(box_title))
        self.play(Create(left_branch), Create(right_branch), Create(hole_left), Create(hole_right), Write(graph_label))
        self.play(Write(notes))

        # Nuqtalarning yaqinlashishi
        self.play(Create(left_dot), Create(right_dot))
        self.play(
            left_dot.animate.move_to(axes.c2p(-0.08, -1)),
            right_dot.animate.move_to(axes.c2p(0.08, 1)),
            run_time=4,
            rate_func=linear
        )
        self.wait(2)

    def on_finish(self):
        super().on_finish(self)
        if os.path.exists("voice_cell2.mp3"):
            a1 = AudioSegment.from_mp3("voice_cell2.mp3")
            a1.export("full_voice2.mp3", format="mp3")

            video_path = self.renderer.file_writer.movie_file_path
            video_clip = VideoFileClip(video_path)
            audio_clip = AudioFileClip("full_voice2.mp3")

            final_clip = video_clip.set_audio(audio_clip)
            final_clip.write_videofile("final_output_BirTomonlamaLimitlar.mp4", codec="libx264", audio_codec="aac")
```

```python
# ==============================================================================
# CELL 3: MAVZU 3 - UZLUKSIZLIK VA UZILISH TURLARI (ML KO'PRIGI)
# ==============================================================================
%%manim -qm -v WARNING UzluksizlikTurlari

from manim import *
from gtts import gTTS
from pydub import AudioSegment
from moviepy.editor import VideoFileClip, AudioFileClip
import os

def generate_voiceover(text, filename):
    tts = gTTS(text=text, lang='tr')
    tts.save(filename)

class UzluksizlikTurlari(Scene):
    def construct(self):
        # 1. Ovoz yaratish
        voice_text = (
            "Uzluksizlik va uzilish turlari. Funksiya uzluksiz bo'lishi uchun uchta shart bajarilishi kerak. "
            "Mashina o'rganishida Sigmoid va ReLU kabi uzluksiz funksiyalar gradient descent ishlashi uchun juda muhimdir."
        )
        generate_voiceover(voice_text, "voice_cell3.mp3")

        # Title
        main_title = Text("3-Mavzu: Uzluksizlik & ML Aktivatsiyalari", font_size=24, color=GREEN_A).to_edge(UP, buff=0.3)
        self.add(main_title)

        # CHAP TOMON: Aktivatsiya funksiyalari (Step vs Sigmoid vs ReLU)
        axes = Axes(
            x_range=[-3, 3, 1],
            y_range=[-0.5, 2.5, 1],
            x_length=5.5,
            y_length=4.5,
            axis_config={"include_numbers": True, "font_size": 16}
        ).to_corner(DL, buff=0.6)

        # Sigmoid function: 1 / (1 + e^-x)
        sigmoid_graph = axes.plot(lambda x: 1 / (1 + np.exp(-x)), x_range=[-3, 3], color=BLUE, stroke_width=3.5)
        sigmoid_label = MathTex("\\sigma(x) = \\frac{1}{1 + e^{-x}}", font_size=22, color=BLUE).move_to(axes.c2p(-1.5, 1.8))

        # ReLU function: max(0, x)
        relu_graph = axes.plot(lambda x: max(0, x), x_range=[-3, 2.3], color=GREEN, stroke_width=3.5)
        relu_label = MathTex("\\text{ReLU}(x) = \\max(0, x)", font_size=22, color=GREEN).move_to(axes.c2p(1.5, 2.2))

        # Step function (Discontinuous)
        step_left = axes.plot(lambda x: 0, x_range=[-3, 0], color=RED, stroke_width=2.5)
        step_right = axes.plot(lambda x: 1, x_range=[0, 3], color=RED, stroke_width=2.5)
        step_label = Text("Step (Uzlikli)", font_size=14, color=RED).move_to(axes.c2p(-1.8, 0.4))

        # O'NG TOMON: Konspekt qutisi
        box = Rectangle(width=5.8, height=5.2, color=GREEN, fill_opacity=0.15).to_corner(DR, buff=0.5)
        box_title = Text("KONSPEKT: UZLUKSIZLIK", font_size=20, weight=BOLD, color=YELLOW).next_to(box.get_top(), DOWN, buff=0.3)

        line1 = Text("UZLUKSIZLIKNING 3 SHARTI:", font_size=15, color=YELLOW)
        line2 = MathTex("1.\\ f(a) \\text{ mavjud}", font_size=20, color=WHITE)
        line3 = MathTex("2.\\ \\lim_{x \\to a} f(x) \\text{ mavjud}", font_size=20, color=WHITE)
        line4 = MathTex("3.\\ \\lim_{x \\to a} f(x) = f(a)", font_size=20, color=GREEN_B)
        line5 = Text("ML DA NEGA KERAK?", font_size=15, color=YELLOW)
        line6 = Text("• Step funksiya: sakrash bor => gradient yo'q!", font_size=13, color=RED_B)
        line7 = Text("• Sigmoid & ReLU: uzluksiz => gradient descent", font_size=13, color=GREEN_A)
        line8 = Text("  muammosiz ishlaydi va o'rganadi.", font_size=13, color=GREEN_A)

        notes = VGroup(line1, line2, line3, line4, line5, line6, line7, line8).arrange(DOWN, aligned_edge=LEFT, buff=0.18).next_to(box_title, DOWN, buff=0.2)

        # Animatsiyalar
        self.play(Create(axes), Create(box), Write(box_title))
        self.play(Write(notes))

        # Step
        self.play(Create(step_left), Create(step_right), Write(step_label))
        self.wait(1)

        # Sigmoid
        self.play(Create(sigmoid_graph), Write(sigmoid_label))
        self.wait(1)

        # ReLU
        self.play(Create(relu_graph), Write(relu_label))
        self.wait(2)

    def on_finish(self):
        super().on_finish(self)
        if os.path.exists("voice_cell3.mp3"):
            a1 = AudioSegment.from_mp3("voice_cell3.mp3")
            a1.export("full_voice3.mp3", format="mp3")

            video_path = self.renderer.file_writer.movie_file_path
            video_clip = VideoFileClip(video_path)
            audio_clip = AudioFileClip("full_voice3.mp3")

            final_clip = video_clip.set_audio(audio_clip)
            final_clip.write_videofile("final_output_UzluksizlikTurlari.mp4", codec="libx264", audio_codec="aac")
```

```python
# ==============================================================================
# CELL 4: MAVZU 4 - HOSILAGA KIRISH: QIYALIKNING LIMITI (NUMERICAL DERIVATIVE)
# ==============================================================================
%%manim -qm -v WARNING HosilaKirish

from manim import *
from gtts import gTTS
from pydub import AudioSegment
from moviepy.editor import VideoFileClip, AudioFileClip
import os

def generate_voiceover(text, filename):
    tts = gTTS(text=text, lang='tr')
    tts.save(filename)

class HosilaKirish(Scene):
    def construct(self):
        # 1. Ovoz yaratish
        voice_text = (
            "Hosilaga kirish. Funksiya qiyaligi sekant chizig'ining limiti orqali topiladi. "
            "h nolga yaqinlashganda qiyalik oltiga teng bo'ladi. Bu gradient descent va modelni o'rgatishning poydevoridir."
        )
        generate_voiceover(voice_text, "voice_cell4.mp3")

        # Title
        main_title = Text("4-Mavzu: Hosila Kirish (Qiyalikning Limiti)", font_size=24, color=ORANGE).to_edge(UP, buff=0.3)
        self.add(main_title)

        # CHAP TOMON: Parabola y = x^2, x=3 nuqtada sekant chizig'i
        axes = Axes(
            x_range=[0, 4.5, 1],
            y_range=[0, 16, 4],
            x_length=5.5,
            y_length=4.5,
            axis_config={"include_numbers": True, "font_size": 16}
        ).to_corner(DL, buff=0.6)

        parabola = axes.plot(lambda x: x**2, x_range=[0, 4.0], color=BLUE, stroke_width=4)
        parabola_label = MathTex("f(x) = x^2", font_size=26, color=BLUE).next_to(axes, UP, buff=0.1)

        # Base point P = (3, 9)
        p_dot = Dot(axes.c2p(3, 9), color=RED, radius=0.09)
        p_label = MathTex("P(3, 9)", font_size=20, color=RED).next_to(p_dot, LEFT, buff=0.15)

        # Tracker for h (h: 1.2 -> 0.01)
        h_tracker = ValueTracker(1.2)

        # Dynamic secant line
        secant_line = always_redraw(lambda: axes.plot(
            lambda x: 9 + ((3 + h_tracker.get_value())**2 - 9) / h_tracker.get_value() * (x - 3),
            x_range=[1.2, 4.2],
            color=YELLOW,
            stroke_width=3
        ))

        q_dot = always_redraw(lambda: Dot(
            axes.c2p(3 + h_tracker.get_value(), (3 + h_tracker.get_value())**2),
            color=YELLOW,
            radius=0.08
        ))

        slope_tex = always_redraw(lambda: MathTex(
            f"h = {h_tracker.get_value():.2f}, \\quad \\text{{Qiyalik}} = {((3+h_tracker.get_value())**2 - 9)/h_tracker.get_value():.2f}",
            font_size=20, color=YELLOW
        ).next_to(axes, DOWN, buff=0.3))

        # O'NG TOMON: Konspekt qutisi
        box = Rectangle(width=5.8, height=5.2, color=ORANGE, fill_opacity=0.15).to_corner(DR, buff=0.5)
        box_title = Text("KONSPEKT: HOSILA URUG'I", font_size=20, weight=BOLD, color=YELLOW).next_to(box.get_top(), DOWN, buff=0.3)

        line1 = MathTex("\\text{Qiyalik} = \\frac{f(3+h) - f(3)}{h}", font_size=22, color=WHITE)
        line2 = MathTex("\\lim_{h \\to 0} \\frac{(3+h)^2 - 9}{h} = 6", font_size=22, color=GREEN_B)
        line3 = Text("1. h -> 0 da 0/0 sirini limit yechadi!", font_size=15, color=WHITE)
        line4 = Text("2. Hosila = Aynan shu nuqtadagi qiyalik.", font_size=15, color=GREEN_A)
        line5 = Text("3. FLOAT TUZOG'I (Python):", font_size=15, color=RED_B)
        line6 = Text("   h = 1e-15 qilsangiz, float xatoligi", font_size=13, color=LIGHT_GRAY)
        line7 = Text("   tufayli qiyalik noto'g'ri chiqadi!", font_size=13, color=LIGHT_GRAY)
        line8 = Text("   Optimal h ≈ 1e-5 deb oling.", font_size=13, color=YELLOW)

        notes = VGroup(line1, line2, line3, line4, line5, line6, line7, line8).arrange(DOWN, aligned_edge=LEFT, buff=0.18).next_to(box_title, DOWN, buff=0.2)

        # Animatsiyalar
        self.play(Create(axes), Create(box), Write(box_title))
        self.play(Create(parabola), Write(parabola_label), Create(p_dot), Write(p_label))
        self.play(Write(notes))
        self.play(Create(secant_line), Create(q_dot), Write(slope_tex))

        # h ni 1.2 dan 0.05 ga tushirish (Sekant urinmaga aylanadi)
        self.play(h_tracker.animate.set_value(0.05), run_time=5, rate_func=linear)
        self.wait(2)

    def on_finish(self):
        super().on_finish(self)
        if os.path.exists("voice_cell4.mp3"):
            a1 = AudioSegment.from_mp3("voice_cell4.mp3")
            a1.export("full_voice4.mp3", format="mp3")

            video_path = self.renderer.file_writer.movie_file_path
            video_clip = VideoFileClip(video_path)
            audio_clip = AudioFileClip("full_voice4.mp3")

            final_clip = video_clip.set_audio(audio_clip)
            final_clip.write_videofile("final_output_HosilaKirish.mp4", codec="libx264", audio_codec="aac")
```
