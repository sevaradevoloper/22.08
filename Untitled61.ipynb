```python
import matplotlib.pyplot as plt
import numpy as np
# ==========================================
# OSON: 1-BO'LIM
# ==========================================

# 1-MASHQ
# Qo'lda hisoblash (w = 0, 1, 2):
# w=0:  pred=[0,0,0,0,0] -> sq_err=[4, 16, 25, 16, 25] -> sum=86  -> C(0) = 86/5 = 17.2
# w=1:  pred=[1,2,3,4,5] -> sq_err=[1, 4, 4, 0, 0]    -> sum=9   -> C(1) = 9/5 = 1.8
# w=2:  pred=[2,4,6,8,10]-> sq_err=[0, 0, 1, 16, 25]  -> sum=42  -> C(2) = 42/5 = 8.4

x1 = np.array([1, 2, 3, 4, 5])
y1 = np.array([2, 4, 5, 4, 5])

def cost1(w):
    return np.mean((w * x1 - y1) ** 2)

for w in [0, 1, 2]:
    print(f"w={w}: C(w)={cost1(w):.1f}")
# Eng kichik cost: w=1 da (C=1.8)


# 2-MASHQ
# Jadval:
# w:         0    1    1.9   2   2.3   3
# Hosila:    -    -     -    0    +    +
# Tushuntirish: w < 2 bo'lganda parabola pastga qiyalanadi (kamayadi), shuning uchun hosila manfiy.
# w = 2 da minimum (eng tubi) joylashgan bo'lib, qiyalik nolga teng.
# w > 2 bo'lganda parabola yuqoriga ko'tariladi (o'sadi), shuning uchun hosila musbat.


# 3-MASHQ
x3 = np.array([2, 4, 6])
y3 = np.array([1, 2, 3])

def cost3(w):
    return np.mean((w * x3 - y3) ** 2)

print(f"C(0.5) = {cost3(0.5):.1f}")
# w=0.5 ideal chiziq, chunki y = 0.5 * x mos keladi va C(0.5) = 0.0 chiqadi.


# ==========================================
# O'RTA: 2-BO'LIM
# ==========================================

# 4-MASHQ
x4 = np.array([2, 3, 4, 5])
y4 = np.array([4, 7, 9, 12])

def gradient4(w):
    return np.mean(2 * x4 * (w * x4 - y4))

# Bashorat: w=0, 1, 2 da manfiy (chunki w minimumdan chapda), w=3 da musbat (minimumdan o'ngda).
for w in [0, 1, 2, 3]:
    print(f"w={w}: gradient={gradient4(w):.1f}")


# 5-MASHQ
def cost4(w):
    return np.mean((w * x4 - y4) ** 2)

w = 0.0
lr = 0.02
for step in range(8):
    g = gradient4(w)
    w = w - lr * g
    print(f"{step+1}: w={w:.4f}, cost={cost4(w):.4f}")


# 6-MASHQ
# lr = 0.005
w_slow = 0.0
for step in range(6):
    w_slow = w_slow - 0.005 * gradient4(w_slow)
    print(f"lr=0.005 step {step+1}: w={w_slow:.4f}, cost={cost4(w_slow):.4f}")

# lr = 0.15
w_fast = 0.0
for step in range(6):
    w_fast = w_fast - 0.15 * gradient4(w_fast)
    print(f"lr=0.15 step {step+1}: w={w_fast:.4f}, cost={cost4(w_fast):.4f}")

# Tushuntirish: lr=0.005 juda kichik bo'lgani uchun qadamlar o'ta sekin va minimumga yetib borish uchun juda ko'p vaqt talab qiladi.
# lr=0.15 juda katta bo'lgani uchun minimum nuqtadan sakrab o'tib ketadi va cost oshib, tebranish hosil qiladi.


# ==========================================
# QIYIN: 3-BO'LIM
# ==========================================

# 7-MASHQ
w7 = 0.0
lr7 = 0.02
tolerance = 0.0001
prev_cost = float('inf')
step7 = 0

while True:
    step7 += 1
    g = gradient4(w7)
    w7 = w7 - lr7 * g
    cur_cost = cost4(w7)
    if abs(prev_cost - cur_cost) < tolerance:
        print(f"To'xtagan qadam: {step7}, w={w7:.4f}, cost={cur_cost:.4f}")
        break
    prev_cost = cur_cost


# 8-MASHQ
x8 = np.array([1, 2, 3, 4, 5, 6])
y8 = np.array([5, 8, 10, 15, 17, 20])

def cost8(w, b):
    return np.mean((w * x8 + b - y8) ** 2)

def grad_w(w, b):
    return np.mean(2 * x8 * (w * x8 + b - y8))

def grad_b(w, b):
    return np.mean(2 * (w * x8 + b - y8))

w8 = 0.0
b8 = 0.0
lr8 = 0.01
cost_history = []

for step in range(300):
    gw = grad_w(w8, b8)
    gb = grad_b(w8, b8)
    w8 = w8 - lr8 * gw
    b8 = b8 - lr8 * gb
    cost_history.append(cost8(w8, b8))

print(f"Yakuniy: w={w8:.2f}, b={b8:.2f}, cost={cost_history[-1]:.2f}")

plt.plot(cost_history)
plt.xlabel("Qadamlar")
plt.ylabel("Cost")
plt.title("Cost funksiyasining pasayishi")
plt.show()

# Izoh: w ≈ 3.14 bu har 1 birlik reklama byudjeti ko'payganda kunlik buyurtmalar soni o'rtacha 3.14 ta oshishini bildiradi.
# b ≈ 1.46 bu reklama byudjeti 0 bo'lganda ham kunlik taxminan 1.46 ta (1-2 ta) bazaviy buyurtma tushishini anglatadi.
```
