```python
%%manim -qm -v WARNING LimitTushunchasi
from manim import *
from gtts import gTTS
from pydub import AudioSegment
from moviepy.editor import VideoFileClip, AudioFileClip
import numpy as np
import os

def generate_voiceover(text, filename):
    tts = gTTS(text=text, lang='tr')
    tts.save(filename)

class LimitTushunchasi(Scene):
    def construct(self):
        # ============================================================
        # 1. AUDIOLARNI YARATISH (bo'lim bo'yicha alohida-alohida)
        # ============================================================
        generate_voiceover(
            "Bugun biz limit tushunchasini o'rganamiz. Funksiya f iks teng ikki x plyus bir "
            "ni oling. X uch songa yaqinlashganda, f iks qayerga yaqinlashadi, degan savolni beramiz.",
            "voice1.mp3"
        )
        generate_voiceover(
            "Jadvalga qarang. X ikki nuqta to'qqiz, ikki nuqta to'qqiz to'qqiz va shunga o'xshash "
            "qiymatlar bilan uchga yaqinlashganda, funksiya qiymati yetti songa yaqinlashmoqda. "
            "Bu ikkala tomondan ham, ya'ni chapdan va o'ngdan ham to'g'ri.",
            "voice2.mp3"
        )
        generate_voiceover(
            "Shuning uchun biz limit iks intiladi uchga, ikki iks plyus bir teng yetti, deb yozamiz. "
            "E'tibor bering, bu misolda limit shunchaki funksiyaning uchdagi qiymatiga teng chiqdi, "
            "lekin bu har doim shunday bo'lavermaydi.",
            "voice3.mp3"
        )
        generate_voiceover(
            "Ikkinchi misolga o'tamiz. F iks teng iks kvadrat minus to'rt bo'lingan iks minus ikkiga. "
            "Iks ikkiga teng bo'lganda, maxraj nolga aylanadi va funksiya bu nuqtada aniqlanmagan. "
            "Lekin algebra yordamida iks kvadrat minus to'rtni iks minus ikki ko'paytirilgan iks "
            "plyus ikki ko'rinishida yozish mumkin, shuning uchun iks ikkiga teng bo'lmaganda, "
            "funksiya iks plyus ikkiga teng bo me'yoriy o'rin egallaydi. Demak, iks ikkiga intilganda, limit to'rtga teng, "
            "garchi funksiyaning o'zi ikki nuqtada aniqlanmagan bo'lsa ham.",
            "voice4.mp3"
        )

        # Ovoz uzunliklarini hisoblash (Sinxronizatsiya uchun)
        dur1 = AudioSegment.from_mp3("voice1.mp3").duration_seconds
        dur2 = AudioSegment.from_mp3("voice2.mp3").duration_seconds
        dur3 = AudioSegment.from_mp3("voice3.mp3").duration_seconds
        dur4 = AudioSegment.from_mp3("voice4.mp3").duration_seconds

        # ============================================================
        # 2. CHAP TOMON — VIZUAL GRAFIKA
        # ============================================================
        axes = Axes(
            x_range=[-1, 5, 1],
            y_range=[-2, 12, 2],
            x_length=5.5,
            y_length=5.5,
            axis_config={"include_numbers": True, "font_size": 20},
        ).to_edge(LEFT, buff=0.6)
        axes_labels = axes.get_axis_labels(x_label="x", y_label="f(x)")

        graph = axes.plot(lambda x: 2 * x + 1, x_range=[-1, 5], color=BLUE)
        graph_label = MathTex("f(x) = 2x + 1", font_size=32, color=BLUE).next_to(
            axes, UP, buff=0.15
        )

        target_dot = Dot(axes.c2p(3, 7), color=YELLOW, radius=0.08)
        v_line = DashedLine(axes.c2p(3, 0), axes.c2p(3, 7), color=YELLOW, stroke_width=2)
        h_line = DashedLine(axes.c2p(3, 7), axes.c2p(0, 7), color=YELLOW, stroke_width=2)
        target_label = MathTex("(3,\\,7)", font_size=26, color=YELLOW).next_to(
            target_dot, UR, buff=0.1
        )

        moving_dot = Dot(color=RED, radius=0.09)
        x_tracker = ValueTracker(2.0)
        moving_dot.add_updater(
            lambda d: d.move_to(axes.c2p(x_tracker.get_value(), 2 * x_tracker.get_value() + 1))
        )
        moving_label = always_redraw(
            lambda: MathTex(f"x={x_tracker.get_value():.3f}", font_size=24, color=RED).next_to(
                moving_dot, DOWN, buff=0.2
            )
        )

        self.play(Write(axes), Write(axes_labels), run_time=1.0)
        self.play(Create(graph), Write(graph_label), run_time=1.0)

        # ============================================================
        # 3. O'NG TOMON — KONSPEKT QUTISI
        # ============================================================
        note_box = Rectangle(
            width=5.6, height=6.2, color=GREY_B, fill_color=GREY_E, fill_opacity=0.15
        ).to_edge(RIGHT, buff=0.5)
        note_title = Text("LIMIT NIMA?", font_size=30, color=YELLOW, weight=BOLD).next_to(
            note_box.get_top(), DOWN, buff=0.25
        )

        note_line1 = Text("x soni 3 ga yaqinlashganda,", font_size=22)
        note_line2 = Text("f(x) qaysi songa yaqinlashadi?", font_size=22)
        note_group1 = VGroup(note_line1, note_line2).arrange(DOWN, aligned_edge=LEFT).next_to(
            note_title, DOWN, buff=0.35
        ).align_to(note_box, LEFT).shift(RIGHT * 0.3)

        formula = MathTex(r"\lim_{x \to 3} (2x+1) = 7", font_size=34, color=GREEN).next_to(
            note_group1, DOWN, buff=0.45
        )

        note_line3 = Text("MUHIM: limit uchun f(a)", font_size=20, color=RED)
        note_line4 = Text("ning o'zi kerak emas!", font_size=20, color=RED)
        note_group2 = VGroup(note_line3, note_line4).arrange(DOWN, aligned_edge=LEFT).next_to(
            formula, DOWN, buff=0.5
        ).align_to(note_box, LEFT).shift(RIGHT * 0.3)

        self.play(Create(note_box), Write(note_title), run_time=0.8)
        self.play(FadeIn(note_group1, shift=UP * 0.2), run_time=0.8)

        # ============================================================
        # 4. AUDIO 1 SINXRON ANIMATSIYASI
        # ============================================================
        self.add(moving_dot, moving_label)
        self.play(
            x_tracker.animate.set_value(2.999),
            run_time=max(1.0, dur1 - 2.6),
            rate_func=rate_functions.ease_in_out_sine,
        )
        self.wait(0.5)

        # ============================================================
        # 5. JADVAL VA AUDIO 2 SINXRONIZATSIYASI
        # ============================================================
        table_data = [["2.9", "6.8"], ["2.99", "6.98"], ["2.999", "6.998"], ["3.001", "7.002"]]
        table = Table(
            table_data,x
            col_labels=[Text("x", font_size=22), Text("f(x)", font_size=22)],
            include_outer_lines=True,
        ).scale(0.35).next_to(axes, DOWN, buff=0.3)

        self.play(Create(table), run_time=1.0)

        x_tracker.set_value(4.0)
        self.play(
            x_tracker.animate.set_value(3.001),
            run_time=max(1.0, dur2 - 2.5),
            rate_func=rate_functions.ease_in_out_sine,
        )

        self.play(Create(v_line), Create(h_line), run_time=0.5)
        self.play(FadeIn(target_dot), Write(target_label), run_time=0.5)
        self.wait(0.5)

        # ============================================================
        # 6. FORMULA VA XULOSA (AUDIO 3)
        # ============================================================
        self.play(Write(formula), run_time=1.0)
        self.wait(1.0)

        self.play(FadeOut(moving_dot), FadeOut(moving_label), FadeOut(table))
        self.play(FadeIn(note_group2, shift=UP * 0.2), run_time=1.0)
        self.wait(max(0.5, dur3 - 3.0))

        # ============================================================
        # 7. IKKINCHI MISOL: (x^2-4)/(x-2), x->2 (AUDIO 4)
        # ============================================================
        self.play(
            FadeOut(graph), FadeOut(graph_label), FadeOut(v_line), FadeOut(h_line),
            FadeOut(target_dot), FadeOut(target_label),
            FadeOut(note_group1), FadeOut(formula), FadeOut(note_group2),
        )

        new_title = Text("f(x) = (x²-4)/(x-2)", font_size=26, color=BLUE).next_to(
            axes, UP, buff=0.15
        )

        def hole_func(x):
            return x + 2

        graph2_left = axes.plot(hole_func, x_range=[-1, 1.98], color=BLUE)
        graph2_right = axes.plot(hole_func, x_range=[2.02, 5], color=BLUE)
        hole_dot = Circle(radius=0.08, color=BLUE, fill_opacity=0).move_to(axes.c2p(2, 4))

        limit_note_title = Text("Nega 4?", font_size=26, color=YELLOW, weight=BOLD).next_to(
            note_box.get_top(), DOWN, buff=0.25
        )
        algebra1 = MathTex(r"x^2-4=(x-2)(x+2)", font_size=26).next_to(
            limit_note_title, DOWN, buff=0.4
        ).align_to(note_box, LEFT).shift(RIGHT * 0.3)

        algebra2 = MathTex(r"\Rightarrow f(x)=x+2 \;\; (x\neq 2)", font_size=26).next_to(
            algebra1, DOWN, buff=0.3
        ).align_to(note_box, LEFT).shift(RIGHT * 0.3)

        conclusion = Text(
            "f(2) mavjud emas,\nlekin limit = 4 mavjud!", font_size=22, color=GREEN
        ).next_to(algebra2, DOWN, buff=0.5).align_to(note_box, LEFT).shift(RIGHT * 0.3)

        self.play(Write(new_title), run_time=0.8)
        self.play(Create(graph2_left), Create(graph2_right), run_time=1.2)
        self.play(Create(hole_dot), run_time=0.5)
        self.play(Write(limit_note_title), run_time=0.5)
        self.play(Write(algebra1), run_time=1.0)
        self.play(Write(algebra2), run_time=1.0)
        self.play(Write(conclusion), run_time=1.0)

        self.wait(max(1.0, dur4 - 6.0))

    def on_finish(self):
        super().on_finish(self)
        self._combine_audio_and_video()

    def _combine_audio_and_video(self):
        combined = AudioSegment.empty()
        for vf in ["voice1.mp3", "voice2.mp3", "voice3.mp3", "voice4.mp3"]:
            if os.path.exists(vf):
                combined += AudioSegment.from_mp3(vf)
                combined += AudioSegment.silent(duration=500)
        combined.export("full_voice_limit.mp3", format="mp3")

        # CRITICAL FIX: PosixPath obyektini matnga (str) o'giramiz
        video_path = str(self.renderer.file_writer.movie_file_path)

        video_clip = VideoFileClip(video_path)
        audio_clip = AudioFileClip("full_voice_limit.mp3")

        if audio_clip.duration > video_clip.duration:
            audio_clip = audio_clip.subclip(0, video_clip.duration)

        final_clip = video_clip.set_audio(audio_clip)
        final_clip.write_videofile(
            "final_output_LimitTushunchasi.mp4", codec="libx264", audio_codec="aac", fps=30
        )
```

```python
%%manim -qm -v WARNING BirTomonlamaLimitlar
from manim import *
from gtts import gTTS
from pydub import AudioSegment
from moviepy.editor import VideoFileClip, AudioFileClip
import numpy as np
import os

def generate_voiceover(text, filename):
    tts = gTTS(text=text, lang='tr')
    tts.save(filename)

class BirTomonlamaLimitlar(Scene):
    def construct(self):
        generate_voiceover("Bir tomonlama limitlar. Modul iks bo'lingan iks funksiyasini ko'raylik. Chapdan minus bir, o'ngdan plyus bir.", "v2_1.mp3")
        generate_voiceover("Chap va o'ng limitlar bir xil emas. Shuning uchun nol nuqtasida umumiy limit mavjud emas.", "v2_2.mp3")

        dur1 = AudioSegment.from_mp3("v2_1.mp3").duration_seconds
        dur2 = AudioSegment.from_mp3("v2_2.mp3").duration_seconds

        axes = Axes(x_range=[-3, 3, 1], y_range=[-2, 2, 1], x_length=5.5, y_length=5.0, axis_config={"include_numbers": True, "font_size": 20}).to_edge(LEFT, buff=0.6)
        axes_labels = axes.get_axis_labels(x_label="x", y_label="f(x)")

        left_branch = axes.plot(lambda x: -1, x_range=[-3, -0.02], color=RED)
        right_branch = axes.plot(lambda x: 1, x_range=[0.02, 3], color=GREEN)
        hole_left = Circle(radius=0.08, color=RED, fill_opacity=0).move_to(axes.c2p(0, -1))
        hole_right = Circle(radius=0.08, color=GREEN, fill_opacity=0).move_to(axes.c2p(0, 1))

        box = Rectangle(width=5.6, height=5.5, color=GREY_B, fill_color=GREY_E, fill_opacity=0.15).to_edge(RIGHT, buff=0.5)
        box_title = Text("CHAP VA O'NG LIMIT", font_size=24, color=YELLOW, weight=BOLD).next_to(box.get_top(), DOWN, buff=0.25)

        l1 = MathTex(r"\lim_{x \to 0^-} \frac{|x|}{x} = -1", font_size=26, color=RED)
        l2 = MathTex(r"\lim_{x \to 0^+} \frac{|x|}{x} = +1", font_size=26, color=GREEN)
        l3 = Text("Chap limit != O'ng limit", font_size=20, color=ORANGE)
        l4 = Text("=> Limit MAVJUD EMAS!", font_size=22, color=RED)

        notes = VGroup(l1, l2, l3, l4).arrange(DOWN, aligned_edge=LEFT, buff=0.35).next_to(box_title, DOWN, buff=0.3).align_to(box, LEFT).shift(RIGHT * 0.3)

        self.play(Write(axes), Write(axes_labels), Create(box), Write(box_title))
        self.play(Create(left_branch), Create(right_branch), Create(hole_left), Create(hole_right))
        self.play(Write(notes), run_time=max(1.0, dur1 + dur2 - 3.0))
        self.wait(1.0)

    def on_finish(self):
        super().on_finish(self)
        combined = AudioSegment.empty()
        for vf in ["v2_1.mp3", "v2_2.mp3"]:
            if os.path.exists(vf):
                combined += AudioSegment.from_mp3(vf) + AudioSegment.silent(duration=500)
        combined.export("full_voice_2.mp3", format="mp3")

        video_path = str(self.renderer.file_writer.movie_file_path)
        video_clip = VideoFileClip(video_path)
        audio_clip = AudioFileClip("full_voice_2.mp3")

        if audio_clip.duration > video_clip.duration:
            audio_clip = audio_clip.subclip(0, video_clip.duration)

        final_clip = video_clip.set_audio(audio_clip)
        final_clip.write_videofile("final_output_02_BirTomonlamaLimitlar.mp4", codec="libx264", audio_codec="aac", fps=30)
        video_clip.close()
        audio_clip.close()
```

```python
%%manim -qm -v WARNING HosilaKirish
from manim import *
from gtts import gTTS
from pydub import AudioSegment
from moviepy.editor import VideoFileClip, AudioFileClip
import numpy as np
import os

def generate_voiceover(text, filename):
    tts = gTTS(text=text, lang='tr')
    tts.save(filename)

class HosilaKirish(Scene):
    def construct(self):
        # 1. AUDIOLARNI YARATISH
        generate_voiceover(
            "Hosila - bu funksiya o'zgarishining oniy tezligidir. Geometrik jihatdan, u kesuvchi chiziqning "
            "h nolga intilgandagi urinmaga aylanish limitidir. Bu tushuncha Machine Learning-da Gradient Descent "
            "algoritmining eng asosiy poydevori hisoblanadi.",
            "v4_1.mp3"
        )
        dur1 = AudioSegment.from_mp3("v4_1.mp3").duration_seconds

        # 2. CHAP TOMON — GRAFIK VA URINMA
        axes = Axes(
            x_range=[0, 4, 1],
            y_range=[0, 16, 4],
            x_length=5.5,
            y_length=5.5,
            axis_config={"include_numbers": True, "font_size": 20}
        ).to_edge(LEFT, buff=0.6)
        axes_labels = axes.get_axis_labels(x_label="x", y_label="f(x)")

        parabola = axes.plot(lambda x: x**2, x_range=[0, 3.8], color=BLUE)
        graph_label = MathTex("f(x) = x^2", font_size=30, color=BLUE).next_to(axes, UP, buff=0.15)

        h_tracker = ValueTracker(1.2)

        # Kesuvchi / Urinma chizig'i
        secant = always_redraw(lambda: axes.plot(
            lambda x: 9 + ((3 + h_tracker.get_value())**2 - 9) / h_tracker.get_value() * (x - 3),
            x_range=[1, 4], color=YELLOW
        ))

        # x=3 nuqta
        dot = Dot(axes.c2p(3, 9), color=RED)
        dot_label = MathTex("(3, 9)", font_size=22, color=RED).next_to(dot, UR, buff=0.05)

        # 3. O'NG TOMON — TUSHUNARLI KONSPEKT QUTISI
        box = Rectangle(
            width=5.8, height=6.2, color=GREY_B, fill_color=GREY_E, fill_opacity=0.15
        ).to_edge(RIGHT, buff=0.4)

        box_title = Text("HOSILA NIMA?", font_size=26, color=YELLOW, weight=BOLD).next_to(
            box.get_top(), DOWN, buff=0.25
        )

        # Konspekt matnlari (Tartibli va tushunarli joylashuv)
        desc1 = Text("1. O'zgarish tezligi:", font_size=20, color=WHITE)
        formula = MathTex(
            r"f'(x) = \lim_{h \to 0} \frac{f(x+h)-f(x)}{h}",
            font_size=26,
            color=GREEN
        )

        desc2 = Text("2. Geometrik ma'nosi:", font_size=20, color=WHITE)
        geom_text = Text("Urinma chiziqning qiyalik burchagi", font_size=18, color=LIGHT_GRAY)

        desc3 = Text("3. ML dagi o'rni:", font_size=20, color=WHITE)
        ml_text = Text("Gradient Descent orqali\nmodel xatosini kamaytirish", font_size=18, color=ORANGE)

        # Bloklarni birlashtirish va tekislash
        content = VGroup(
            desc1, formula,
            desc2, geom_text,
            desc3, ml_text
        ).arrange(DOWN, aligned_edge=LEFT, buff=0.25).next_to(
            box_title, DOWN, buff=0.35
        ).align_to(box, LEFT).shift(RIGHT * 0.3)

        # ANIMATSIYALAR
        self.play(Write(axes), Write(axes_labels), Create(parabola), Write(graph_label), run_time=1.0)
        self.play(Create(box), Write(box_title), run_time=0.8)
        self.play(FadeIn(content), FadeIn(dot), Write(dot_label), run_time=1.2)

        # Animatsiyani ovozga sinxronlash (h qiymati 0 ga intiladi)
        self.play(Create(secant))
        self.play(
            h_tracker.animate.set_value(0.001),
            run_time=max(2.0, dur1 - 3.0),
            rate_func=rate_functions.ease_in_out_sine
        )
        self.wait(1.0)

    def on_finish(self):
        super().on_finish(self)
        combined = AudioSegment.empty()
        if os.path.exists("v4_1.mp3"):
            combined += AudioSegment.from_mp3("v4_1.mp3") + AudioSegment.silent(duration=500)
        combined.export("full_voice_4.mp3", format="mp3")

        # Video path str ga o'tkaziladi (PosixPath xatosini oldini olish uchun)
        video_path = str(self.renderer.file_writer.movie_file_path)
        video_clip = VideoFileClip(video_path)
        audio_clip = AudioFileClip("full_voice_4.mp3")

        if audio_clip.duration > video_clip.duration:
            audio_clip = audio_clip.subclip(0, video_clip.duration)

        final_clip = video_clip.set_audio(audio_clip)
        final_clip.write_videofile(
            "final_output_04_HosilaKirish.mp4", codec="libx264", audio_codec="aac", fps=30
        )
        video_clip.close()
        audio_clip.close()
```
