```python
import seaborn as sns
from sklearn.linear_model import LinearRegression

df = sns.load_dataset("anscombe")

for name,g in df.groupby("dataset"):
  X = g[["x"]]; y = g["y"]
  m = LinearRegression().fit(X,y)
  print(name,round(m.coef_[0],2),
    round(m.score(X,y),3))
sns.lmplot(data=df,x="x",y="y",
           col="dataset",col_wrap=2)
```

df.groupby("dataset") — Datasetni ichidagi dataset turlariga (I, II, III, IV) ajratib chiqadi (ya'ni 4 ta alohida guruhga bo'ladi).

X = g[["x"]] va y = g["y"] — Har bir guruh uchun kirish ma'lumoti (x) va natija (y) ustunlarini ajratib oladi.

LinearRegression().fit(X, y) — O'sha guruh uchun chiziqli regressiya modelini o'qitadi (w qiyalik va b siljishni topadi).
print(...) — Har bir datasetning nomini (name),
qiyaligini (coef_[0]) verguldan keyin 2 xona qilib va R
2
  aniqligini (score) verguldan keyin 3 xona qilib ekranga chiqaradi.
E'tibor bergan bo'lsangiz, bu tsiklni ishga tushirsangiz, to'rtala boshqacha dataset uchun ham bir xil (yoki deyarli bir xil) koeffitsiyent va R

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

df = pd.read_csv("/content/house-prices-advanced-regression-techniques/train.csv")

df.shape
```

```python
feats = [
    'GrLivArea', 'OverallQual','GarageCars','TotalBsmtSF',   'YearBuilt'
]

X = df[feats]
y = df['SalePrice']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

model = LinearRegression().fit(X_train, y_train)

pred = model.predict(X_test)
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    median_absolute_error,
    mean_absolute_percentage_error
)

pred = model.predict(X_test)

print("MAE:", mean_absolute_error(y_test, pred))
print("MSE:", mean_squared_error(y_test, pred))
print("R2:", r2_score(y_test, pred))
print("MedAE:", median_absolute_error(y_test, pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, pred)))
print("MAPE:", mean_absolute_percentage_error(y_test, pred))
```
